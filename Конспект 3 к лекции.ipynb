# Знакомство с рекурсией, декораторами и генераторами

## Рекурсия

**Функция** - это блок кода с именем и аргументами, который может вернуть объект в качестве результата. 

````python
def foo():
    pass
````

Функции могут вызывать другие функции, например, определим функцию **power**, которая возводит число `a` в степень `b`, и вторую функцию **hypotenuse**, которая вычисляет гипотенузу прямоугольного треугольника с катетами `a` и `b`, и при этом вызывает функцию **power** три раза. 

````python
def power(a,b):
    return a ** b

def hypotenuse(a,b):
    c = power(a, 2) + power(b, 2)
    return power(c, 0.5)

print(hypotenuse(3, 4))
````

**Рекурсия** - это поведение функции, при котором она вызывает сама себя.
При использовании рекурсивных функций **ОБЯЗАТЕЛЬНО** необходимо указать условие остановки внутри функции, которое называтся **базовым случаем**. Так, если не добавить условие остановки функции - это приведёт к зацикливанию функции: 

````python
def func():
    func()
func()
````

Каждый рекурсивный вызов функции занимает место в специальной области памяти - стеке вызовов. При каждом новом вызове в стек добавляется информация о текущем состоянии функции (аргументы, локальные переменные). Таким образом ошибка **RecursionError: maximum recursion depth exceeded** возникает, когда рекурсивная функция выполняет слишком много вложенных вызовов, превышая установленный лимит глубины рекурсии, который по умолчанию обычно составляет 1000. Эта ошибка является защитным механизмом, предотвращающим переполнение стека вызовов. Обычный бесконечный цикл `while True` она не остановит, так как он не создаёт новых вызовов и будет работать, пока его не прервут вручную.

При помощи **рекурсии** можно заменить циклы **for** и **while**.
Для начала создадим функцию, которая будет поочередно выводить элементы списка:

````python
def loop(a):
    for x in a:
        print(x)

loop([1,2,3])
````

Теперь переопределим эту функцию для случая использования рекурсии, и начнем с **базового случая**, когда список пуст и с ним нечего делать. В таком случае мы просто завершаем выполнение функции при помощи `return`. В рекурсивном случае выводим первый элемент списка a[0] и сделаем рекурсивный вызов со списком без первого числа. Так список будет становиться меньше, пока не окажется пустым, после чего выполнится **базовый случай**.

````python
def loop(a):
    if not a: # базовый случай
        return
    print(a[0])
    loop(a[1:])

loop([1,2,3])
````

### Дополнение: рекурсия может возвращать результат

Рекурсивная функция не обязана только выводить текст. Чаще всего она **возвращает** ответ,
собирая его из двух частей: «первый элемент» и «действие для остальных». Это тот же
шаблон, что в `loop`, только вместо `print` сложение.

````python
def total(a):
    if not a:                    # базовый случай: сумма пустого списка
        return 0
    return a[0] + total(a[1:])   # первый + сумма остальных

print(total([5, 3, 4]))          # 5 + (3 + (4 + 0)) = 12
````

> Каждый срез `a[1:]` создаёт новый список, поэтому на длинных списках такая рекурсия
> будет выполняться медленно, а на списке длиннее лимита глубины закончит свое выполнение с ошибкой `RecursionError`.

Рассмотрим как можно заменить цикл **while**. Для начала создадим функцию, которая будет выполнять обратный отсчет от заданного пользователем числа **countdown**.

````python
def countdown(n):
    while n != 0:
        print(n)
        n -= 1
countdown(5)
````

Теперь переопределим эту функцию для случая использования рекурсии, и опять же начнем с базового случая, когда число равно нулю, считать дальше некуда, поэтому завершаем выполнение функции при помощи `return`. В рекурсивном случае выводим число и выполняем рекурсивный вызов функции **countdown** с числом уменьшенным на единицу.

````python
def countdown(n):
    if n == 0: # базовый случай
        return
    print(n)
    return countdown(n-1)
countdown(5)
````

### Дополнение: что будет с отрицательными числами?

`countdown(-1)` никогда не дойдёт до нуля: `-1, -2, -3, ...`. Версия с `while`
зациклится навсегда, рекурсивная закончит свое выполнение с ошибкой `RecursionError`. Поэтому базовый случай
надёжнее писать с запасом - `n <= 0`, так как он срабатывает и на нуле, и на всём, что ниже.

````python
def countdown_safe(n):
    if n <= 0:          # базовый случай с запасом
        return
    print(n)
    countdown_safe(n - 1)

countdown_safe(3)
countdown_safe(-1)      # ничего не печатает и не падает
````

Теперь рассмотрим классическую задачу для применения рекурсии: вычисление факториала числа. Факториалом числа является произведение всех натуральных чисел от 1 до заданного числа. Решим задачу без применения рекурсии.

````python
def factorial(n):
    x = 1
    for i in range (1, n + 1):
        x *= i
    return x

print(factorial(5))
````

Переопределим эту функцию для использования рекурсии, и опять же начнем с базового случая:
* 0! = 1, 
* 1! = 1.

Факториал любого другого числа будет являться произведением этого числа на факториал этого числа уменьшенного на 1. Так, постепенно уменьшая число, мы придём к базовому случаю и выполним выход из функции. 

````python
def factorial(n):
    if n <= 1: # базовый случай
        return 1
    return n * factorial(n - 1)

print(factorial(5))
````

Теперь рассмотрим применение рекурсивной функции для вычисления n-го числа последовательности Фибоначчи. Это последовательность, в которой:
* первые два числа равны 0 и 1;
* каждое последующее - сумме двух предыдущих:

Решим эту задачу при помощи рекурсии:

````python
def fibonacci(n):
    if n < 2: # базовый случай: n = 0 или n = 1
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(35))
````

После запуска программы можно заметить, что вычисление происходит не моментально. Причина в том, что при каждом вызове `fib(n)` функция вызывает саму себя дважды: один раз для `fib(n - 1)` и один раз для `fib(n - 2)`. Это приводит к тому, что одни и те же значения пересчитываются много раз. Получается ветвящееся рекурсивное дерево, в котором одни и те же подзадачи повторяются снова и снова.

Чтобы увидеть это дерево, сначала разберёмся, как вообще выполняются рекурсивные вызовы, на простом примере без ветвления. Давайте посмотрим, как работает стек вызовов для `factorial(3)`:

**Что происходит в памяти:**

1. `factorial(3)` -> ждет результат `factorial(2)`
2. `factorial(2)` -> ждет результат `factorial(1)`
3. `factorial(1)` -> базовый случай, возвращает 1
4. `factorial(2)` получает 1 -> возвращает 2
5. `factorial(3)` получает 2 -> возвращает 6

При этом каждый "ожидающий" вызов занимает место в стеке до тех пор, пока не получит результат от следующего вызова.

### Дополнение: как можно увидеть стек вызовов?

Добавим в функцию параметр `depth`, который будет являться глубиной вызова, и будем выводить отступ.

````python
def factorial_verbose(n, depth=0):
    print("    " * depth + f"factorial({n}) вызван")
    if n <= 1:
        result = 1
    else:
        result = n * factorial_verbose(n - 1, depth + 1)
    print("    " * depth + f"factorial({n}) вернул {result}")
    return result

factorial_verbose(3)
````

### Дополнение: дерево вызовов Фибоначчи

У факториала каждый вызов порождает **один** следующий и получается цепочка вызовов.
У Фибоначчи каждый вызов порождает **два** и получается дерево вызовов. Ниже показан пример для `fib(5)`:

15 вызовов ради одного маленького числа, и `fib(3)` посчитан дважды, `fib(2)` - трижды.
Посчитаем вызовы для больших `n`. Счётчик будем хранить в словаре: функция **изменяет** словарь,
а не присваивает новое значение имени, поэтому изменение видно снаружи.

````python
CALLS = {"fib": 0}

def fib(n):
    CALLS["fib"] += 1
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)

for n in [5, 10, 15, 20, 25]:
    CALLS["fib"] = 0
    value = fib(n)
    print(f"fib({n:>2}) = {value:>6}   вызовов: {CALLS['fib']:>7}")
````

> Каждые пять шагов `n` примерно в 11 раз больше вызовов. Для `fib(35)` их почти
> 30 миллионов, отсюда и заметная пауза. В разделе про декораторы мы ускорим эту
> функцию, **не изменяя ни строчки в ней самой**.

Теперь сравним решение с применением цикла **for**, которое выполняет вычисление 100-ого члена Фибоначчи моментально:

````python
def fibonacci(n):
    if n <= 1:
        return n
    a, b = 0, 1
    for i in range(2, n + 1):
        a, b = b, a + b
    return b

print(fibonacci(100))
````

Поэтому старайтесь избегать написания рекурсивных функций, которые умножают число вызовов за счёт того, что вызывают себя многократно. 

**Когда же использовать рекурсию?**
* Для задач, которые естественно рекурсивны (обход деревьев, комбинаторика)
* Когда глубина рекурсии предсказуемо мала
* Когда читаемость кода важнее производительности

### Дополнение: задача, которая естественно рекурсивна

Каталог новостей: рубрика может содержать подрубрики, те, в свою очередь свои подрубрики,
а внизу находятся списки заголовков. Сколько уровней, заранее неизвестно, поэтому
написать нужное число вложенных циклов `for` невозможно. Рекурсия решает это
одним правилом: «если это список, вот ответ; если словарь, сделай то же самое
для каждого вложенного». `isinstance(x, list)` проверяет, является ли `x` списком.

````python
CATALOG = {
    "наука": {
        "биология": ["Учёные открыли новый вид бактерий"],
        "медицина": {"вирусы": ["Новый вирус обнаружен", "Вакцина прошла испытания"]},
    },
    "спорт": ["Сборная выиграла матч", "Тренер покинул клуб"],
}

def count_items(node):
    if isinstance(node, list):        # базовый случай: дошли до списка
        return len(node)
    total = 0
    for child in node.values():       # шаг: то же самое для каждой подрубрики
        total += count_items(child)
    return total

print(count_items(CATALOG))
print(count_items(CATALOG["наука"]))
````

> Функции всё равно, сколько уровней вложенности: «вирусы» лежат на три уровня
> глубже, чем «спорт», и это ничего не изменило. Так же устроены JSON-ответы
> веб-сервисов, деревья разбора предложений и папки на диске.

## Декораторы

Иногда нужно изменить поведение функции, например, добавить логирование, измерение времени выполнения или кеширование результатов. Но менять сам код функции при этом **не хочется** (или **нельзя**). В таких случаях удобно использовать **декораторы**.

**Декоратор** - это специальная конструкция, которая оборачивает функцию и может изменить её поведение до, после или вместо обычного вызова.

Итак, как же выглядит реализация простого декоратора?

Начнём со свойств функций, которые пригодятся для понимания, без которых декоратор выглядит как магия.

**1. Функция - это значение.** Имя без скобок - сама функция. Её можно передать
в переменную или другую функцию, а также вернуть из функции.

````python
def shout(text):
    return text.upper() + "!"

action = shout              # без скобок: не вызов, а сама функция
print(action("привет"))

def apply_twice(func, value):
    return func(func(value))

print(apply_twice(shout, "ура"))
````

**2. Вложенная функция запоминает переменные внешней**, даже когда внешняя уже
закончила свое выполнение. Такая функция называется **замыканием**.

````python
def make_greeting(word):
    def greet(name):
        return f"{word}, {name}!"     # word берётся из внешней функции
    return greet

hello = make_greeting("Привет")
bye = make_greeting("Пока")
print(hello("Олег"), bye("Олег"))
````

> Декоратор, похож на такую реализацию: принимает функцию, создаёт внутри
> обёртку, которая помнит исходную функцию, и возвращает обёртку.

````python
def null_decorator(func):
    return func
````

В данном случае `null_decorator` является вызываемым объектом, он принимает на вход другой вызываемый объект и возвращает тот же самый входной объект, не изменяя его.
Используем его для декорирования (или обертывания) другой функции:

````python
def greet():
    return 'Hello!'

greet = null_decorator(greet)

greet()
````

Другой способ записи декорирования - это использование "синтаксического сахара". Python автоматически преобразует запись с `@` в явное присваивание, что делает код чище и читаемее:

````python
@null_decorator
def greet():
    return 'Hello!'

greet()
````

В данном случае декоратор на нашу функцию никак не повлиял.
Теперь давайте создадим декоратор `start_stop`, который будет выводить текст до и после выполнения функции `Hello`:

````python
def start_stop(f):
    def wrapper():
        print("Это текст до выполнения функции")
        f()
        print("Это текст после выполнения функции")
    return wrapper

@start_stop
def Hello():
    name = input("Введите своё имя: ")
    print(f"Привет, {name}")

Hello()
````

Видно, что функция отработала без каких-либо изменений за исключением добавления вывода текста в консоль до выполнения функции и после неё.

> **Обратите внимание.** Эта обёртка подходит только для функции без аргументов,
> которая ничего не возвращает: `wrapper()` не принимает параметров, а результат `f()`
> никуда не передаёт, обёрнутая функция всегда возвращает `None`. Универсальный шаблон
> с `*args`, `**kwargs` и `return`, будет представлен чуть ниже, в примере `log_arguments`.

Теперь давайте создадим функцию, которая повышает регистр введенного пользователем текста, при этом не изменяя функционал функции `input_text`, добавив декоратор `uppercase`: 

````python
def uppercase(func):
    def wrapper():
        original_result = func() # вызываем оригинальную функцию и сохраняем результат в переменную
        modified_result = original_result.upper() # модифицируем результат, полученный от оригинальной функции
        return modified_result # возвращаем модифицированный результат тому, кто вызвал функцию
    return wrapper

@uppercase
def input_text():
    return input("Введите текст: ")

print(input_text())
````

Видно, что функция `input_text`, которая просто возвращает данные, введённые пользователем, отработала с учётом созданной нами обертки без изменения функционала оригинальной функции.

Часто нужно, чтобы декорируемая функция могла принимать аргументы. Для этого используется ***args** и ****kwargs**:
* ***args** - собирает все позиционные аргументы в кортеж,
* ****kwargs** - собирает все именованные аргументы в словарь,
что позволяет декоратору работать с ЛЮБОЙ функцией, независимо от количества и типа аргументов.

````python
def log_arguments(func):
    def wrapper(*args, **kwargs):
        print(f"Вызов функции {func.__name__} с аргументами:")
        print(f"  Позиционные: {args}")
        print(f"  Именованные: {kwargs}")
        return func(*args, **kwargs)  # передаем аргументы дальше
    return wrapper

@log_arguments
def multiply(a, b):
    return a * b

print(multiply(5, 3))
````

### Дополнение: декоратор, который считает количество вызовов

У функции, как и у любого объекта, можно создать **атрибут**: `wrapper.calls = 0`, которая создается во время вызова функции.
Обёртка увеличивает его при каждом вызове. Так можно узнать, сколько раз вызывалась
любая функция, не изменяя её код.

````python
def count_calls(func):
    def wrapper(*args, **kwargs):
        wrapper.calls += 1
        return func(*args, **kwargs)
    wrapper.calls = 0
    return wrapper

@count_calls
def word_count(text):
    return len(text.split())

for text in ["кот спит", "собака бежит по улице", "ёж"]:
    word_count(text)
print("word_count вызвана", word_count.calls, "раза")
````

### Дополнение: возвращаемся к Фибоначчи: кэш-декоратор

**Кэш** - память ответов. Если функцию уже вызывали с такими аргументами,
обёртка возвращает предыдущий ответ и не считает его заново. Ключ словаря - это кортеж
аргументов `args`, значение - результат.

Рассмотрим как это будет с рекурсивной функцией. При этом, код самой функции изменяеть не будем.

````python
def cache(func):
    def wrapper(*args):
        if args not in wrapper.memory:
            wrapper.memory[args] = func(*args)
        return wrapper.memory[args]
    wrapper.memory = {}
    return wrapper

@count_calls
@cache
def fib_fast(n):
    if n < 2:
        return n
    return fib_fast(n - 1) + fib_fast(n - 2)

print("fib_fast(25) =", fib_fast(25), "  вызовов:", fib_fast.calls)
print("fib_fast(35) =", fib_fast(35))
print("fib_fast(100) =", fib_fast(100))
````

> **49 вызовов вместо 242 785**, и `fib(35)` теперь считается мгновенно. Внутри
> `fib_fast` рекурсивные вызовы идут по имени `fib_fast` - то есть через обёртку,
> поэтому каждое значение считается ровно один раз.
>
> В стандартной библиотеке такой декоратор уже есть: `from functools import cache`
> (или `lru_cache`).

### Дополнение: порядок выполнения декораторов

Декораторов над функцией может быть несколько. Применяются они **снизу вверх**:
ближайший к `def` выполняется первым. Запись

```python
@bold
@italic
def title(): ...
```

аналогична `title = bold(italic(title))`.

````python
def bold(func):
    def wrapper(*args, **kwargs):
        return "<b>" + func(*args, **kwargs) + "</b>"
    return wrapper

def italic(func):
    def wrapper(*args, **kwargs):
        return "<i>" + func(*args, **kwargs) + "</i>"
    return wrapper

@bold
@italic
def title_1(text):
    return text

@italic
@bold
def title_2(text):
    return text

print(title_1("Новости"))
print(title_2("Новости"))
````

> В примере с `fib_fast` выше `@count_calls` стоит **над** `@cache`, поэтому считает
> все обращения к функции. Если поставить его после кэша, он посчитает только
> реальные вычисления, те, на которые кэш не ответил сам.

### Дополнение: потерянное имя функции

После декорирования функция является `wrapper`. Имя и документация исходной
функции изменяется и в сообщениях об ошибках и в `help()` вы увидите не то.

````python
@log_arguments
def divide(a, b):
    """Делит a на b."""
    return a / b

print(divide.__name__, "|", divide.__doc__)
````

Изменить это можно копированием имени и документации в обёртку. В стандартной библиотеке
для этого существует декоратор `functools.wraps`, который должен находится перед `wrapper`:

```python
from functools import wraps

def log_arguments(func):
    @wraps(func)                 # копирует __name__, __doc__ и др. из func
    def wrapper(*args, **kwargs):
        ...
    return wrapper
```

Без импорта то же самое делается двумя строками:

````python
def log_arguments_fixed(func):
    def wrapper(*args, **kwargs):
        print(f"Вызов {func.__name__}{args}")
        return func(*args, **kwargs)
    wrapper.__name__ = func.__name__
    wrapper.__doc__ = func.__doc__
    return wrapper

@log_arguments_fixed
def divide(a, b):
    """Делит a на b."""
    return a / b

print(divide.__name__, "|", divide.__doc__)
print(divide(10, 4))
````

Таким образом, декораторы изменяют поведение вызываемого объекта с помощью обертки, поэтому вам не нужно постоянно изменять оригинал. Вызываемый объект не подвергается постоянным изменениям, его поведение изменяется только при декорировании. Это позволяет «присоединять» к существующим функциям повторно используемые модули, такие как логирование и другие инструменты. Именно это делает декораторы такой мощной функцией в Python, которая часто используется в стандартной библиотеке.

## Генераторы

В Python функции могут возвращать не только готовое значение, но и **генератор**, который является специальным объектом, который выдаёт элементы по одному, **по запросу**, что особенно удобно, когда вы работаете с **большим количеством данных** и нет смысла загружать всё в память сразу.  При этом вместо `return` используется ключевое слово `yield`, а его отличительными особенностями являются следующее:
* Приостанавливает выполнение функции, а не завершает её
* Сохраняет состояние всех локальных переменных, а не освобождает все локальные переменные
* Может возвращать значения многократно, а не один раз
* Функция продолжается с места остановки, а не сначала.

Вызов такой функции **не выполняет её тело**, а только создаёт объект-генератор. Значения из него получают встроенной функцией `next()`: каждый вызов `next(gen)` выполняет тело функции до следующего `yield` и возвращает значение. Когда значения заканчиваются, выводится исключение `StopIteration`. 

````python
def normal_function():
    numbers = [1, 2, 3]
    return numbers
print("Функция с return")
print(normal_function())

def generator_function():
    numbers = [1, 2, 3]
    for num in numbers:
        yield num

print("Генератор с yield")
gen = generator_function()
print(next(gen)) 
print(next(gen))
print(next(gen))
print(next(gen))
````

> Ошибка `StopIteration` в последней строке, **ожидаема**: значений было три,
> а `next` вызван в четвёртый раз. Цикл `for` сам останавливается,
> поэтому с генераторами обычно работают через `for`.

### Дополнение

Добавим `print` внутрь генератора. Если бы он работал как обычная функция,
все сообщения «выдаю» появились бы сразу при вызове. На деле каждое появляется
только тогда, когда значение **запросили**.

````python
def numbers_loud():
    for i in [1, 2, 3]:
        print(f"  генератор: выдаю {i}")
        yield i
    print("  генератор: значения кончились")

gen = numbers_loud()
print("генератор создан, но ещё ничего не выдал")
print("получил", next(gen))
print("сделаю что-то ещё...")
print("получил", next(gen))
````

Генераторы идеально подходят для использования в циклах `for`:

````python
def number_generator(n):
    for i in range(n):
        yield i

for num in number_generator(5):
    print(num)
````

### Дополнение: генератор читается один раз

Генератор - это **поток**, а не хранилище, прочитанное значение уже не вернуть.
На этом спотыкаются чаще всего, ошибки нет, просто второй результат неверный.

````python
gen = number_generator(5)
print("первая сумма:", sum(gen))     # 0+1+2+3+4 = 10
print("вторая сумма:", sum(gen))     # 0: генератор уже прочитан

values = list(number_generator(5))   # если нужен многократно, нужно сохранить его в список
print(sum(values), sum(values))
````

### Дополнение: генераторное выражение

Короткий генератор можно записать одной строкой, как списковое включение,
только в **круглых** скобках. Квадратные скобки сразу создает весь список в памяти,
круглые выдают значения по одному. Если выражение - это единственный аргумент функции,
вторые скобки можно не писать.

````python
texts = ["Кот спит на диване", "Собака бежит по улице", "Ёж"]

lengths_list = [len(t.split()) for t in texts]     # список: всё сразу
lengths_gen = (len(t.split()) for t in texts)      # генератор: по одному

print(lengths_list)
print(lengths_gen)                                 # не числа, а объект-генератор
print(sum(lengths_gen))
print(max(len(t) for t in texts))                  # вторые скобки не нужны
````

### Дополнение: конвейер из генераторов

Генераторы можно соединять в цепочку: каждый берёт значения у предыдущего
и передаёт дальше. Так обчыно обрабатывают большие данные, которые не помещаются в память:
одновременно в работе находится **одно** слово, а не весь текст. А если нужное
найдено раньше, остальное даже не читается.

````python
def read(texts):
    for text in texts:
        print(f"  читаю: {text!r}")
        yield text

def lower(stream):
    for text in stream:
        yield text.lower()

def words(stream):
    for text in stream:
        for word in text.split():
            yield word

corpus = ["Сборная выиграла матч", "Учёные открыли вирус", "Курс доллара вырос"]

for word in words(lower(read(corpus))):
    print(word)
    if word == "открыли":
        break        # третий текст так и не прочитан
````

**Когда использовать генераторы**

* Если работаете с большими объемами данных
* Необходимо обрабатывать данные по частям, а не целиком
* Для экономии используемой памяти
* Тогда, когда данные можно вычислять в момент вызова

**Когда не стоит использовать генераторы**
* Если нужен случайный доступ к элементам
* Требуется многократное использование результата
* Данные небольшие и проще работать со списком
