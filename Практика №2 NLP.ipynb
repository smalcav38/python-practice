## Задача 1. Первый цикл

*Новое: `for ... in ...`*

Цикл читается словами: «**для каждого** элемента **из** списка - сделай то,
что написано ниже с отступом».

```python
for животное in ["кот", "пёс"]:
    print(животное)
```

`животное` — это имя, которое вы придумываете сами. На каждом шаге цикла
в него кладётся очередной элемент списка.

````python
animals = ["кот", "пёс", "ёж"]

for animal in animals:
    print(animal)
````

Три строки вывода из двух строк кода. Список из тысячи элементов
дал бы тысячу строк - код при этом не изменился бы ни на символ.

## Задача 2. Отступ решает всё

*Новое: отступ как часть языка*

В большинстве языков тело цикла обводят скобками. В Python его обозначает
**отступ** - четыре пробела слева. Всё, что с отступом, повторяется;
первая же строка без отступа выполняется один раз, после цикла.

Отступ здесь не оформление, а синтаксис. Сдвинули строку - изменили программу.

**Посмотрите внимательно: две последние
строки кода отличаются только отступом.**

````python
animals = ["кот", "пёс", "ёж"]

for animal in animals:
    print("внутри цикла:", animal)
    print("   эта строка тоже внутри")
print("а эта — уже снаружи, она одна")
````

**Ваша очередь:** для первых четырёх рубрик напечатайте строку

*Новое: `range`*

Иногда нужен не сам элемент, а его **номер**. `range(n)` даёт номера
от `0` до `n - 1` - правая граница снова не входит.

- `range(3)` -> 0, 1, 2
- `range(2, 5)` -> 2, 3, 4
- `range(0, 10, 3)` -> 0, 3, 6, 9

````python
for i in range(3):
    print(i)

print()
for i in range(2, 5):
    print(i)

print()
print(len(TITLES))
for i in range(3):
    print(i, TITLES[i])
````

## Задача 5. Счётчик: копим сумму

*Новое: `+=` внутри цикла*

Самый частый приём в обработке данных: завести переменную **до** цикла,
а внутри прибавлять к ней по чуть-чуть.

`total += x` - это короткая запись для `total = total + x`.

Важно: строка `total = 0` стоит **до** цикла, без отступа. Если поставить её
внутрь, она будет сбрасываться на каждом шаге, и в конце останется последнее
значение вместо суммы.

````python
numbers = [10, 20, 30]

total = 0
for number in numbers:
    total += number
    print("шаг:", number, "накоплено:", total)

print("итого:", total)
````

**Ваша очередь:** посчитайте циклом, сколько всего символов и сколько
всего слов во всех десяти текстах. (Проверить себя легко: `sum` вы уже знаете.)

**Должно получиться:**

```

слов всего: 256

```

````python
# Задача 5
total_chars = 0
total_words = 0

for text in TEXTS:
    total_chars += len(text)
    total_words += len(text.split())

print("символов всего:", total_chars)
print("слов всего:", total_words)
print("в среднем слов в новости:", total_words / len(TEXTS))
````

## Задача 7. `enumerate` - элемент и его номер

*Новое: `enumerate`*

Когда нужен и элемент, и его номер, пишут `enumerate`. Он выдаёт **пару**:
сначала номер, потом элемент - и оба сразу раскладываются по двум именам.

`start=1` заставляет нумерацию начинаться с единицы - для людей, а не для Python.

````python
animals = ["кот", "пёс", "ёж"]

for i, animal in enumerate(animals):
    print(i, animal)

print()
for i, animal in enumerate(animals, start=1):
    print(i, animal)
````

## Задача 8. `zip` - два списка сразу

*Новое: `zip`*

Наши списки идут параллельно. Чтобы пройти по ним вместе, есть `zip`:
он на каждом шаге берёт по одному элементу из каждого списка.

Это надёжнее, чем цикл по номерам: не собьётесь на единицу, и `zip` сам
остановится на более коротком списке.

**Предскажите вывод и запустите.**

````python
animals = ["кот", "пёс", "ёж"]
sounds = ["мяу", "гав", "фыр"]

for animal, sound in zip(animals, sounds):
    print(animal, "говорит", sound)

print()
print("zip останавливается на коротком:")
for a, b in zip([1, 2, 3], ["а", "б"]):
    print(a, b)
````

**Ваша очередь:** напечатайте первые пять новостей в виде

```
наука | Ёмкость аккумулятора выросла
```

## Задача 9. Таблица

*Новое: ничего. Это сборка задач 6–8*

`enumerate` и `zip` можно объединить. Читается так: «пронумеруй пары».
Скобки вокруг `(a, b)` обязательны - иначе Python не поймёт, где номер,
а где пара.

Чтобы столбики сошлись, строку дополняют пробелами: `.ljust(n)` - до ширины `n`
слева направо, `.rjust(n)` - наоборот.

````python
animals = ["кот", "пёс"]
sounds = ["мяу", "гав"]

for i, (animal, sound) in enumerate(zip(animals, sounds), start=1):
    print(str(i).rjust(2), animal.ljust(6), sound)
````

# Часть 2. Условия

Цикл делает одно и то же с каждым элементом. Условие позволяет делать
**разное** - в зависимости от того, что за элемент попался.

## Задача 10. `if`

*Новое: `if`, сравнения*

`if` читается «если». Если условие истинно - выполняется то, что ниже
с отступом. Если нет - просто пропускается.

Условие - это выражение, которое даёт `True` или `False`. Сравнения:
`==` (равно), `!=` (не равно), `>`, `<`, `>=`, `<=`.

**Двойное равно в сравнении - не опечатка.** Одинарное `=` кладёт значение
в переменную, двойное `==` сравнивает. Перепутать их - классическая ошибка.

**Предскажите вывод и запустите.**

````python
word = "бактерия"

if len(word) > 5:
    print("длинное слово")

if word == "кот":
    print("это кот")

print("конец")
````

Второе условие ложно, и его строка просто не выполнилась -
никакой ошибки при этом нет.

**Ваша очередь:** для первого текста напечатайте «длинная новость»,
если в нём больше 180 символов, и «рубрика - спорт», если первая рубрика
равна `"спорт"`.

**Должно получиться:**

```
длинная новость
рубрика - спорт
```

````python
# Задача 10

if len(TEXTS[0]) > 180:
    print("длинная новость")

if TOPICS[0] == "спорт":
    print("рубрика - спорт")
````

## Задача 11. `if` / `else`

*Новое: `else`*

`else` - «иначе». Выполнится ровно одна из двух веток, никогда обе и никогда ни одной.

````python
for word in ["кот", "бактерия"]:
    if len(word) > 5:
        print(word, "— длинное")
    else:
        print(word, "— короткое")
````

## Задача 12. `if` / `elif` / `else`

*Новое: `elif`*

Когда вариантов больше двух, между `if` и `else` вставляют `elif` -
сокращение от «else if», «а иначе если».

Условия проверяются сверху вниз, и **срабатывает первое подходящее**,
остальные не проверяются. Поэтому порядок веток важен: если поставить
`< 30` выше `< 25`, вторая ветка не сработает никогда.

````python
for count in [12, 27, 40]:
    if count < 20:
        print(count, "мало")
    elif count < 30:
        print(count, "средне")
    else:
        print(count, "много")
````

**Ваша очередь:** пройдите по текстам и присвойте каждому ярлык

## Задача 24. Группировка

*Новое: словарь списков*

Задача, которая в анализе данных встречается чаще любой другой: разложить
записи по группам.

Приём тот же, что у частотного словаря, только копится не число, а список.
И раз в словаре ещё нет ключа, его надо сначала создать с пустым списком:

```python
if ключ not in groups:
    groups[ключ] = []
groups[ключ].append(значение)
```

````python
pairs = [("животные", "кот"), ("птицы", "сова"), ("животные", "ёж")]

groups = {}
for kind, name in pairs:
    if kind not in groups:
        groups[kind] = []
    groups[kind].append(name)

print(groups)

for kind, names in groups.items():
    print(kind.ljust(10), len(names), ", ".join(names))
````

> **\*** У словарей есть готовый метод для этого приёма: `groups.setdefault(ключ, []).append(значение)`
> делает то же самое одной строкой. Перепишите своё решение через него
> и убедитесь, что результат не изменился.

````python
# Задача 24
groups = {}
for topic, title in zip(TOPICS, TITLES):
    if topic not in groups:
        groups[topic] = []
    groups[topic].append(title)

for topic, titles in groups.items():
    print(topic.ljust(10), '-', len(titles))
    for title in titles:
        print('  ', title)

````

## Задача 25. `match` / `case`

*Новое: `match`*

Когда вариантов много, цепочка `if / elif / elif / else` становится длинной,
и в ней повторяется одно и то же имя. Для таких случаев есть `match`:

```python

        ...
    case образец_2:
        ...
    case _:
        ...
```

Значение сверху по очереди сравнивается с образцами, срабатывает **первый
подходящий**. `case _` - «всё остальное», его пишут последним.
Несколько вариантов в одной ветке перечисляют через `|`.

Ничего такого, чего нельзя написать через `if`, `match` не даёт - он даёт
читаемость.

> **Осторожно.** `case topic:` без кавычек - это **не** сравнение с переменной
> `topic`, а создание новой переменной с таким именем. Такая ветка сработает
> всегда и «съест» все остальные. Для сравнения пишите само значение: `case "спорт":`.

````python
for topic in ["спорт", "наука", "погода"]:
    match topic:
        case "спорт":
            print(topic, "-> отдел спорта")
        case "наука" | "технологии":
            print(topic, "-> научный отдел")
        case _:
            print(topic, "-> рубрика неизвестна")
````

**Ваша очередь:** разнесите все рубрики ленты по отделам —
спорт в «спорт», экономика и Россия в «политика и экономика»,
наука и технологии в «наука». Заодно посчитайте словарём, сколько новостей
досталось каждому отделу.

**Должно получиться:**

```
спорт                  2
политика и экономика   4
наука                  4
```

````python
# Задача 25
dept_counts = {}
for topic in TOPICS:
    match topic:
        case "спорт":
            dept = "спорт"
        case "экономика" | "Россия":
            dept = "политика и экономика"
        case "наука" | "технологии":
            dept = "наука"
        case _:
            dept = "рубрика неизвестна"
    dept_counts[dept] = dept_counts.get(dept, 0) + 1

for dept, count in dept_counts.items():
    print(dept.ljust(20), count)

````

## Задача 26. Классификатор по ключевым словам

*Новое: ничего. Это всё сразу*

Вот и первая осмысленная задача обработки текста: угадать рубрику новости,
не зная её, - только по словам в тексте.

Способ самый грубый: список слов-примет для каждой рубрики. Дальше вы
познакомитесь с методами получше, но начинается всё отсюда - и работает
это лучше, чем кажется.

Заодно вы посчитаете **точность** - долю верных ответов. Считать её нужно
только по тем новостям, где классификатор вообще дал ответ: если «не знаю»
засчитывать в ошибки, метрика будет врать в одну сторону, а если в успехи -
в другую.

````python
demo_texts = ["Тренер покинул клуб", "Ставка снижена до 15%", "Дождь заливает город"]
demo_topics = ["спорт", "экономика", "погода"]

for text, topic in zip(demo_texts, demo_topics):
    lowered = text.lower()
    if "тренер" in lowered or "матч" in lowered:
        guess = "спорт"
    elif "ставка" in lowered or "оборот" in lowered:
        guess = "экономика"
    else:
        guess = "не знаю"
    print(topic.ljust(10), "->", guess)
````

**Ваша очередь:** напишите такой классификатор для трёх рубрик -
спорт, экономика, наука - и примените ко всем десяти новостям.
Для каждой напечатайте настоящую рубрику, угаданную и пометку
ВЕРНО / ОШИБКА / ПРОПУСК. В конце выведите три числа и точность
на распознанных новостях.

**Должно получиться:**

```
[0] спорт       -> спорт       ВЕРНО
[1] спорт       -> спорт       ВЕРНО
[2] экономика   -> экономика   ВЕРНО
[3] экономика   -> экономика   ВЕРНО
[4] наука       -> наука       ВЕРНО
[5] наука       -> наука       ВЕРНО
[6] технологии  -> не знаю     ПРОПУСК
[7] технологии  -> не знаю     ПРОПУСК
[8] Россия      -> не знаю     ПРОПУСК
[9] Россия      -> не знаю     ПРОПУСК

Верно: 6   Ошибок: 0   Не знаю: 4
Точность на распознанных: 100% (посчитана по 6 новостям)
```

> Четыре пропуска - это новости о технологиях и о России, для которых
> вы не задали ни одного слова-приметы. Ошибок при этом ноль: классификатор
> честно молчит там, где не знает. Так и должно быть - молчать лучше, чем врать.
>
> **\*** Добавьте слова-приметы для технологий и для России так, чтобы
> пропусков осталось не больше двух. И проверьте, не сломались ли при этом
> прежние ответы: почти наверняка какое-нибудь новое слово встретится
> и в чужой рубрике.

````python
# Задача 26

verno = oshibok = neznaju = 0

for i, (text, topic) in enumerate(zip(TEXTS, TOPICS)):
    lowered = text.lower()

    if "тренер" in lowered or "матч" in lowered:
        guess = "спорт"
    elif "ставка" in lowered or "оборот" in lowered:
        guess = "экономика"
    elif "учёные" in lowered or "ученые" in lowered:
        guess = "наука"
    else:
        guess = "не знаю"

    if guess == topic:
        mark = "ВЕРНО"
        verno += 1
    elif guess == "не знаю":
        mark = "ПРОГУСК"
        neznaju += 1
    else:
        mark = "ОШИБКА"
        oshibok += 1

    print(f"[{i}]", topic.ljust(10), "->", guess.ljust(10), mark)

raspoznano = verno + oshibok
print()
print(f"Верно: {verno}  Ошибок: {oshibok}  Не знаю: {neznaju}")
print(f"Точность на распознанных: {verno / raspoznano:.0%} (посчитана по {raspoznano} новостям)")
````

## Задача 27. Консоль новостей

*Новое: `int()`, разбор команды*

Финальная задача: маленькая программа, которая понимает команды.

Две вещи, которых ещё не было:

- `int("4")` превращает строку в число. Из команды `show 4` вы получаете
  строку `"4"`, а номером списка может быть только число.
- `line.split()` разрезает команду на части: первая - само слово команды,
  вторая (если есть) - аргумент. Проверять `len(parts) > 1` обязательно,
  иначе на голой команде `show` программа упадёт.

Чтобы тетрадь не зависала, команды берутся из готового списка.
Настоящий ввод с клавиатуры - в задании со звёздочкой.

````python
COMMANDS = ["stats", "show 1", "hello"]

for line in COMMANDS:
    print(">", line)
    parts = line.split()
    command = parts[0]

    match command:
        case "stats":
            print("новостей:", len(TEXTS))
        case "show":
            number = int(parts[1])
            print(TITLES[number])
        case _:
            print("Команды: stats | show N")
    print()
````

**Ваша очередь:** сделайте консоль, понимающую пять команд:

- `stats` - сколько новостей, рубрик и словоупотреблений;
- `show N` - рубрика и заголовок новости номер N;
- `topic рубрика` - заголовки новостей этой рубрики;
- `find слово` - заголовки новостей, в тексте которых встречается это слово
  (поиск без учёта регистра);
- `quit` - выйти из цикла;
- любое другое - подсказка со списком команд.

Обязательно предусмотрите `show` без номера и `show` с номером вне диапазона -
программа не должна падать.

Проверьте её на этом списке команд:

```python
COMMANDS = ["stats", "show 4", "topic технологии", "find тренер",
            "show", "show 99", "привет", "quit", "show 0"]
```

**Должно получиться:**

```
> stats
Новостей: 10  Рубрик: 5  Слов: 256

> show 4
[наука] Ёмкость аккумулятора выросла

> topic технологии
В рубрике «технологии» новостей: 2
   Вышел патч безопасности
   Оператор запустил сеть

> find тренер
Найдено: 2
   Сборная вышла в четвертьфинал
   Тренер покинул клуб

> show
Ошибка: команде show нужен номер. Пример: show 3

> show 99
Ошибка: номер должен быть от 0 до 9

> привет

> quit

```

> Последняя команда `show 0` не выполнилась: `break` вышел из цикла
> на `quit`, и до неё дело не дошло. Так и задумано.
>
> **\*** Замените список команд на настоящий ввод: `while True:`
> и `line = input("> ")`. Запускайте такую ячейку **последней** -
> пока программа ждёт ввода, тетрадь занята. Выход - команда `quit`.

````python
# Задача 27

# функция подготовки слов: нижний регистр, знаки и цифры -> пробел,
# остаются только буквенные слова длиннее двух символов
def words_of(text):
    s = text.lower()
    for sign in ",.!?;":
        s = s.replace(sign, " ")
    for digit in "0123456789":
        s = s.replace(digit, " ")
    return [w for w in s.split() if w.isalpha() and len(w) > 2]

# считаем слова по всей ленте: заголовки + тексты
all_words = []
for title, text in zip(TITLES, TEXTS):
    all_words += words_of(title) + words_of(text)

while True:
    line = input("> ").strip()
    parts = line.split()
    cmd = parts[0] if parts else ""

    if cmd == "stat":
        print("Новостей:", len(TEXTS), " Рубрик:", len(set(TOPICS)), " Слов:", len(all_words))
    elif cmd == "show":
        if len(parts) < 2:
            print("Ошибка: команде show нужен номер. Пример: show 3")
        elif not parts[1].isdigit() or int(parts[1]) >= len(TITLES):
            print("Ошибка: номер должен быть от 0 до", len(TITLES) - 1)
        else:
            n = int(parts[1])
            print(f"[{n}]", TITLES[n])
    elif cmd == "topic":
        rubrica = " ".join(parts[1:])
        titles = [t for t, topic in zip(TITLES, TOPICS) if topic == rubrica]
        print(f'В рубрике "{rubrica}" новостей: {len(titles)}')
        for t in titles:
            print("  ", t)
    elif cmd == "find":
        word = " ".join(parts[1:]).lower()
        found = [t for t, text in zip(TITLES, TEXTS) if word in text.lower()]
        print("Найдено:", len(found))
        for t in found:
            print("  ", t)
    elif cmd == "quit":
        print("До встречи.")
        break
    else:
        print("Подсказка: stat | show N | topic рубрика | find слово | quit")

````

| --- | --- |
| сделать с каждым | `for x in список:` |
| тело цикла | отступ в 4 пробела |
| по символам строки | `for ch in строка:` |
| по номерам | `for i in range(n):` |
| накопить сумму | `total = 0` до цикла, `total += x` внутри |
| накопить список | `res = []` до цикла, `res.append(x)` внутри |
| номер и элемент | `for i, x in enumerate(список, start=1):` |
| два списка сразу | `for a, b in zip(A, B):` |
| номер и пара | `for i, (a, b) in enumerate(zip(A, B)):` |
| если | `if условие:` |
| иначе если / иначе | `elif условие:` / `else:` |
| сравнение | `==`, `!=`, `>`, `<`, `>=`, `<=` |
| и / или / не / входит | `and`, `or`, `not`, `in` |
| пропустить шаг | `continue` |
| прекратить цикл | `break` |
| словарь | `{"ключ": значение}` |
| значение по ключу | `d["ключ"]`, ошибка если ключа нет |
| безопасно, с умолчанием | `d.get("ключ", 0)` |
| есть ли ключ | `"ключ" in d` |
| цикл по парам | `for k, v in d.items():` |
| все значения | `d.values()`, `sum(d.values())` |
| частотный словарь | `d[x] = d.get(x, 0) + 1` |
| словарь списков | `if k not in d: d[k] = []` затем `d[k].append(v)` |
| топ по частоте | список кортежей `(частота, слово)` + `.sort(reverse=True)` |
| много вариантов | `match x:` / `case "a" \| "b":` / `case _:` |
| строка в число | `int("4")` |
| выравнивание | `.ljust(n)`, `.rjust(n)` |
