## Что такое функция

**Функция** - это именованный блок кода, который выполняет определённую задачу
и может быть вызван из других частей программы. Функции помогают избежать
дублирования кода, делают программу структурированной и читаемой.

Преимущества:

- **Повторное использование.** Написав функцию один раз, её можно вызывать многократно.
- **Модульность.** Программу, разделённую на функции, проще разрабатывать.
- **Простота отладки.** Ошибку легче найти в небольшом блоке кода.
- **Читаемость.** Функция с понятным именем объясняет, что делает код.

## Определение и вызов

Функция определяется ключевым словом `def`, за которым следуют имя функции,
список параметров в круглых скобках и двоеточие. Тело функции пишется
**с отступом** относительно строки `def`.

```python
def имя_функции(параметры):
    """Строка документации (docstring)."""
    # тело функции
    ...
```

Определение само по себе ничего не выполняет, оно только создаёт функцию.
Код внутри сработает, когда функцию **вызовут** по имени со скобками.

````python
def hello():
    """Выводит приветственное сообщение."""
    print("Здравствуйте! Добро пожаловать в мир Python.")

hello()
````

Функция с параметром:

````python
def greet(name):
    print(f"Привет, {name}!")

greet("Стив Джобс")
greet("Дорогой студент")
greet("Гвидо ван Россум")
````

- **`def`** - ключевое слово, объявляющее функцию;
- **`greet`** - имя функции;
- **`name`** - параметр функции;
- **`print(f"Привет, {name}!")`** - тело функции.

Строку документации (docstring), в которой будет созданное Вами описание, можно прочитать через `help()`:

````python
help(hello)
````

## Встроенные функции

В Python существует множество функций, доступных без всякого импорта:

- **`len()`** - длина объекта;
- **`print()`** - вывод на экран;
- **`sum()`** - сумма элементов;
- **`type()`** - тип объекта;
- **`min()`, `max()`, `sorted()`, `round()`, `abs()`** и другие.

````python
print(len("Hello"))   # 5
print(type(10))       # <class 'int'>
print(sum([1, 2, 3])) # 6
````

## Параметры и аргументы

Эти два слова часто путают, но это разные вещи:

- **Параметры** - имена в заголовке функции, в круглых скобках после `def`.
- **Аргументы** - конкретные значения, которые передаются при вызове.

В строке `def greet_user(name, age):` параметры - `name` и `age`.
В строке `greet_user("Иван", 25)` аргументы - `"Иван"` и `25`.

````python
def greet_user(name, age):
    print(f"Привет, {name}! Тебе {age} лет.")

greet_user("Иван", 25)
````

### Позиционные аргументы

Передаются **в порядке следования** параметров: первый аргумент - первому
параметру, второй - второму. Перепутали порядок - получили другой результат,
и Python об этом не предупредит.

````python
def user_info(name, age, city):
    """Выводит информацию о пользователе."""
    print(f"Имя: {name}, возраст: {age}, город: {city}")

user_info("Иван", 25, "Томск")
user_info(25, "Томск", "Иван")   # ошибки нет — есть неверный результат
````

### Именованные аргументы

При вызове можно указать, какому параметру соответствует значение:
`параметр=значение`. Тогда порядок не важен, а вызов читается без просмотра
в определение функции.

Именованные аргументы полезны, когда параметров много. Позиционные можно
сочетать с именованными, но **позиционные всегда идут первыми**.

````python
user_info(age=25, city="Москва", name="Иван")
user_info("Иван", city="Томск", age=19)

# раскомментируйте - позиционный после именованного нельзя:
# user_info(name="Иван", 25, "Томск")
````

### Значения по умолчанию

Параметру можно задать значение, которое используется, если аргумент
не передан. Параметры со значением по умолчанию пишутся **после** обязательных.

````python
def greet(name, message="Добро пожаловать"):
    print(f"{message}, {name}!")

greet("Алиса")                  # Добро пожаловать, Алиса!
greet("Боб", "Привет")          # Привет, Боб!
greet(name="Чарли", message="Здравствуйте")
````

````python
def hello(name="Гость"):
    """Выводит приветствие с именем."""
    print(f"Здравствуйте, {name}!")

hello("Алиса")
hello()
````

## Возврат значения: `return`

`return` возвращает значение из функции и **сразу завершает** её - строки
после `return` не выполняются.

Главное отличие от `print`: `print` показывает значение человеку, `return`
отдаёт его программе. Результат `return` можно присвоить переменной,
сложить, передать в другую функцию. Результат `print` - нельзя.

````python
def rectangle_area(width, height):
    """Возвращает площадь прямоугольника."""
    return width * height

area = rectangle_area(5, 3)
print(area)                          # 15
print(rectangle_area(2, 2) + area)   # 19 — результат можно использовать дальше
````

````python
def add(a, b):
    return a + b

result = add(5, 3)
print(result)   # 8
````

### Функция без `return` возвращает `None`

Если в функции нет `return` (или он без значения), она возвращает особое
значение `None` или «ничего».

````python
def add_and_print(a, b):
    print(a + b)          # печатает, но не возвращает

result = add_and_print(2, 3)
print("result =", result)  # None

# раскомментируйте — с None складывать нельзя:
# print(result + 1)
````

### Несколько значений

Через запятую после `return` можно вернуть несколько значений, которые будут являться
кортежем. Его удобно сразу распаковывать.

````python
def get_full_name(first_name, last_name):
    return first_name, last_name

first, last = get_full_name("Джон", "Доу")
print(first)
print(last)
print(get_full_name("Джон", "Доу"))   # на самом деле это кортеж
````

## Переменное число аргументов: `*args` и `**kwargs`

Иногда количество аргументов заранее неизвестно.

- **`*args`** собирает все лишние **позиционные** аргументы в кортеж.
- **`**kwargs`** собирает все лишние **именованные** аргументы в словарь.

Имена `args` и `kwargs` - соглашение; важны звёздочки.

### `*args`

````python
def sum_numbers(*args):
    """Возвращает сумму всех переданных чисел."""
    print("  внутри функции args =", args)
    return sum(args)

print(sum_numbers(1, 2, 3))        # 6
print(sum_numbers(5, 10, 15, 20))  # 50
````

### `**kwargs`

````python
def book_info(**kwargs):
    """Выводит информацию о книге."""
    for key, value in kwargs.items():
        print(f"{key}: {value}")

book_info(title="Война и мир", author="Лев Толстой", year=1869)
# title: Война и мир
# author: Лев Толстой
# year: 1869
````

Ключами словаря становятся **имена аргументов**, как они написаны в вызове.
Поэтому `year=1869` превращается в ключ `"year"`, а не в «год».

## Глобальные и локальные переменные

- **Глобальные перемнные** определены вне всех функций. Внутри функции их можно **читать**.
- **Локальные переменные** созданы внутри функции и существуют только во время её работы.

````python
x = 10  # глобальная

def func():
    y = 5  # локальная
    print(f"y = {y}")
    print(f"x = {x}")   # чтение глобальной - можно

func()
print(f"x = {x}")

# раскомментируйте - y снаружи не существует:
# print(y)
````

### Чтение и присваивание в одной функции

Если внутри функции есть присваивание имени, Python считает это имя
**локальным во всей функции** - даже в строках выше присваивания. Попытка
сначала прочитать его, а потом изменить, даёт `UnboundLocalError`.

````python
counter = 0

def increment():
    counter = counter + 1   # Python считает counter локальной — а её ещё нет
    return counter

# раскомментируйте:
# increment()
````

### `global`

Ключевое слово `global` разрешает функции изменять глобальную переменную.
Работает, но пользоваться им стоит редко: функция, которая может изменять
что-то снаружи, непредсказуема, и найти в какой момент времени перменная была изменена затруднительно.

Зачастую лучше передавать значение параметром, а возвращать результат через `return`.

````python
x = 10

def set_to_20():
    global x
    x = 20

set_to_20()
print(x)   # 20

# так лучше — функция ничего не знает о глобальных именах:
def increased(value):
    return value + 1

x = increased(x)
print(x)   # 21
````

## Как аргументы передаются в функцию

В Python аргумент передаётся **по ссылке на объект**, т.е. параметр внутри функции -
это ещё одно имя для **того же самого** объекта, который передали. Копии
не создаётся.

Что будет видно снаружи функции, зависит не от типа, а от того, **что делает функция**:

- **изменяет сам объект** (`.append()`, `.remove()`, `d[k] = v`) - изменение
  видно снаружи функции, ведь объект один;
- **присваивает параметру новое значение** (`x = ...`, `x += ...` для чисел
  и строк) - внутри функции имя начинает указывать на новый объект,
  а снаружи ничего не меняется.

Числа, строки и кортежи неизменяемы - у них есть только второй вариант.
Поэтому кажется, что «неизменяемые не меняются в функции». Но и список
не изменится, если его не менять, а переприсвоить.

### Неизменяемый объект: переприсваивание

````python
def change_number(num):
    num += 10          # num теперь указывает на новое число 15
    print(f"внутри: {num}")

x = 5
change_number(x)
print(f"снаружи: {x}")   # 5
````

````python
def change_str(my_str):
    my_str += " мир"   # создаётся новая строка

greeting = "Привет"
change_str(greeting)
print(greeting)   # Привет
````

### Изменяемый объект: изменение

````python
def add_elem(lst):
    lst.append(4)      # меняется сам список

my_list = [1, 2, 3]
add_elem(my_list)
print(my_list)   # [1, 2, 3, 4]
````

### Изменяемый объект: переприсваивание

Список изменяемый, но функция его не изменяет - 
она создаёт новый список и привязывает к нему локальное имя.

````python
def rebind(lst):
    lst = lst + [4]    # новый список, исходный не тронут

my_list = [1, 2, 3]
rebind(my_list)
print(my_list)   # [1, 2, 3]
````

> **Практическое правило.** Функция, которой передали список или словарь,
> может его изменить. Если функция должна вернуть изменённые данные -
> пусть создаёт новый объект и возвращает его через `return`, а переданный
> оставляет без изменений.

## Ловушка изменяемого значения по умолчанию

Значение по умолчанию вычисляется **один раз**, когда Python выполняет строку `def`.
Если это список или словарь, все вызовы получают **один и тот же** объект,
и он копит данные между вызовами.

````python
def add_item(item, bag=[]):     # ловушка
    bag.append(item)
    return bag

print(add_item("а"))   # ['а']
print(add_item("б"))   # ['а', 'б'] - список из прошлого вызова никуда не делся

def add_item_safe(item, bag=None):
    if bag is None:
        bag = []
    bag.append(item)
    return bag

print(add_item_safe("а"))
print(add_item_safe("б"))
````

## Лямбда-функции

**Лямбда** - маленькая безымянная функция из одного выражения. Результат
выражения возвращается автоматически, `return` не пишется.

```python
lambda аргументы: выражение
```

Отличия от `def`:

- **Определение**: `lambda` вместо `def`.
- **Размер**: только одно выражение, без нескольких строк и присваиваний.
- **Имя**: лямбда анонимна.

Присвоить лямбду переменной можно, но так почти не пишут: если функции
нужно имя, её пишут через `def`.

````python
add = lambda x, y: x + y      # работает, но лучше def
print(add(5, 3))

def add(x, y):                 # то же самое, читается лучше
    return x + y
print(add(5, 3))
````

### Где лямбды нужны на самом деле: `key=`

Главное место лямбды там, где функцию нужно **передать** другой функции.
Чаще всего - в параметр `key` у `sorted`, `min`, `max`: он говорит, по какому
признаку сравнивать элементы.

````python
counts = [("новость", 5), ("учёные", 12), ("вирус", 7)]

print(sorted(counts, key=lambda pair: pair[1]))
print(sorted(counts, key=lambda pair: pair[1], reverse=True))
print(max(counts, key=lambda pair: pair[1]))

words = ["кот", "бактерия", "вирус"]
print(sorted(words, key=len))   # готовую функцию можно передать и без лямбды
````

### `map`

`map(функция, последовательность)` применяет функцию к каждому элементу.

````python
numbers = [1, 2, 3, 4]

squared = list(map(lambda x: x ** 2, numbers))
print(squared)                      # [1, 4, 9, 16]

print([x ** 2 for x in numbers])    # то же самое включением
````

## Генераторы (кратко)

Генераторная функция использует `yield` вместо `return`. При вызове она
не выполняется сразу, а возвращает объект-генератор; значения используются
по одному, когда их запрашивает цикл.

````python
def count_up_to(max_value):
    count = 1
    while count <= max_value:
        yield count
        count += 1

for num in count_up_to(5):
    print(num)
````

Генераторное выражение похоже на списковое включение, но в круглых скобках.
Оно не хранит всю последовательность в памяти, и **прочитать его можно только
один раз**.

````python
squared = (x ** 2 for x in range(5))
print(sum(squared))   # 30
print(sum(squared))   # 0 — генератор уже прочитан
````

## Итог

- **Функция** - именованный блок кода; определяется через `def`, выполняется при вызове.
- **Параметры** - имена в определении, **аргументы** - значения при вызове.
- Аргументы бывают **позиционные** и **именованные**; у параметров могут быть **значения по умолчанию**.
- **`return`** отдаёт значение программе; без него функция возвращает **`None`**.
- **`*args`** и **`**kwargs`** принимают переменное число аргументов.
- **Локальные** переменные существуют внутри функции; **глобальные** лучше не изменять внутри функции.
- Аргумент передаётся **по ссылке на объект**: изменение объекта видно снаружи, переприсваивание - нет.
- Изменяемое **значение по умолчанию** - ловушка; пишите `None` и создавайте объект внутри.
- **`lambda`** - короткая функция; главное её место - параметр `key=`.
