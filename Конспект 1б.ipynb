Рассмотрим три конструкции:

- **цикл** - сделать одно и то же с каждым;
- **условие** - сделать разное в зависимости от того, что попалось;
- **словарь** - считать и группировать по ключу.

Цикл, внутри которого условие - это девяносто процентов всей обработки
данных. Остальное - библиотеки, которые делают то же самое, только гораздо быстрее.

````python
TEXTS = [
    "Учёные открыли новый вид бактерий в глубинах Тихого океана",
    "Сборная выиграла матч со счётом 3:1",
    "Новый вирус обнаружен учёными в лаборатории Новосибирска",
]

LABELS = ["наука", "спорт", "наука"]
SCORES = [5, 3, 4]

print("записей:", len(TEXTS))
````

# 1. Цикл `for`

Чтобы напечатать три записи, раньше нужно было выводить три строки.
Для тысячи записей понадобилась бы тысяча. Как упростить такой вывод?

Если цикл описать простыми словами, то  «**для каждого** элемента **из** последовательности -
сделай то, что ниже с отступом».

````python
for text in TEXTS:
    print(text)
````

Три строки вывода из двух строк кода - и код не изменится, даже если записей
миллион. Имя `text` вы придумываете сами; на каждом шаге в него кладётся
очередной элемент.

## Отступ - это синтаксис, а не оформление

В большинстве языков тело цикла обводят фигурными скобками. В Python его
обозначает **отступ**: четыре пробела слева. Всё, что с отступом, повторяется;
первая строка без отступа выполняется один раз, после цикла.

Сдвинули строку - изменили программу. Забыли отступ - `IndentationError`.

````python
for label in LABELS:
    print("внутри:", label)
    print("   тоже внутри")
print("снаружи, выполнилось один раз")
````

## `for` работает не только со списком

По строке цикл идёт посимвольно, по результату `.split()` - по словам.

````python
for letter in "кот":
    print(letter)

print()
print(TEXTS[1])
for word in TEXTS[1].split():
    print(word, len(word))
````

## `range` - цикл по номерам

Когда нужен не элемент, а его номер. `range(n)` даёт числа от `0` до `n - 1` -
правая граница снова не входит.

- `range(3)` - 0, 1, 2
- `range(2, 5)` - 2, 3, 4
- `range(0, 10, 3)` - 0, 3, 6, 9

````python
for i in range(3):
    print(i, LABELS[i], SCORES[i])

print()
for i in range(0, 10, 3):
    print(i)
````

# 2. Накопление

Второй по важности приём после самого цикла. Переменная заводится **до** цикла,
а внутри к ней прибавляется по чуть-чуть.

`total += x` - короткая запись для `total = total + x`.

**Строка `total = 0` стоит до цикла, без отступа.** Если поставить её внутрь,
она будет сбрасываться на каждом шаге, и в конце останется последнее значение
вместо суммы. Ошибки при этом не будет - будет неправильное число.

````python
print(TEXTS)
total = 0
for text in TEXTS:
    total += len(text.split())
    print("шаг:", len(text.split()), "накоплено:", total)

print("слов всего:", total)
````

Тот же приём, но копится не число, а список: пустой список до цикла,
`.append()` внутри. Так из одного списка получают другой - и это основа
почти любой обработки.

````python
lengths = []
for text in TEXTS:
    lengths.append(len(text.split()))

print(lengths)
print(sum(lengths), min(lengths), max(lengths))
````

# 3. `enumerate` и `zip`

**`enumerate`** выдаёт пару «номер, элемент» - оба сразу раскладываются
по двум именам. `start=1` включает нумерацию с единицы.

**`zip`** проходит по нескольким спискам одновременно. Наши списки параллельны,
поэтому `zip` - основной способ их обходить: он надёжнее цикла по номерам,
потому что не даёт ошибиться на единицу, и сам останавливается на более
коротком списке.

````python
for i, text in enumerate(TEXTS, start=1):
    print(i, text[:30])

print()
for label, score in zip(LABELS, SCORES):
    print(label, score)

print()
print(list(zip([1, 2, 3], ["а", "б"])))
````

Их можно объединить: «пронумеруй пары». Скобки вокруг `(a, b)` обязательны -
иначе Python не поймёт, где номер, а где пара. Это как **распаковка**
из прошлого занятия, просто написанная в заголовке цикла.

````python
for i, (label, score) in enumerate(zip(LABELS, SCORES), start=1):
    print(str(i).rjust(2), label.ljust(8), score)
````

# 4. Условия

`if` читается «если»: истинно - выполняется то, что ниже с отступом;
ложно - просто пропускается, без всякой ошибки.

`else` - «иначе». `elif` - «а иначе если», для третьего и последующих вариантов.

**Условия проверяются сверху вниз, и срабатывает первое подходящее.**
Порядок веток поэтому важен: поставьте `< 50` выше `< 30`, и вторая ветка
не сработает никогда.

````python
for text in TEXTS:
    count = len(text.split())
    if count < 7:
        label = "короткая"
    elif count < 9:
        label = "обычная"
    else:
        label = "длинная"
    print(count, label)
````

## Из чего собирают условие

Сравнения: `==`, `!=`, `>`, `<`, `>=`, `<=`.
**Двойное равно - не опечатка:** одинарное `=` присваивает значение, двойное
сравнивает.

Соединяют словами: `and` - оба истинны, `or` - хотя бы одно, `not` - наоборот.
`x in y` - «встречается ли»: в строке ищет подстроку, в списке - элемент.

Значение типа `bool` уже готовое условие, сравнивать его с `True` не нужно:
пишут `if verified:`, а не `if verified == True:`.

````python
text = TEXTS[0]

print(len(text) > 30 and "учёные" in text.lower())
print("учёные" in text)
print("учёные" in text.lower())
````

> Вторая и третья строки различаются только `.lower()` - и дают разный ответ,
> потому что в тексте слово с заглавной буквы. Приводить к нижнему регистру
> перед поиском нужно почти всегда; забыть об этом - обычное дело.

## Цикл плюс условие

Две самые частые пары во всей обработке данных: **посчитать подходящие**
и **отобрать подходящие в новый список**.

````python
long_texts = 0
for text in TEXTS:
    if len(text.split()) > 6:
        long_texts += 1
print("длинных:", long_texts)
print(f"доля: {long_texts / len(TEXTS):.0%}")

print()
science = []
for text, label in zip(TEXTS, LABELS):
    if label == "наука":
        science.append(text)
print("научных:", len(science))
````

# 5. `continue`, `break` и `while`

**`continue`** - «этот элемент мне не нужен, переходи к следующему».
**`break`** - «выйти из цикла целиком».

`break` ставят, когда нужное найдено и продолжать бессмысленно: зачем
проверять оставшиеся тысячу восемьсот записей, если ответ найден на третьей.

````python
for text in TEXTS:
    if len(text.split()) < 7:
        continue
    print("длинная:", text[:35])

print()
for i, label in enumerate(LABELS):
    print("смотрю", i, label)
    if label == "спорт":
        print("нашёл на позиции", i)
        break
````

## `while` - пока верно условие

`for` проходит по готовой последовательности. `while` работает, пока условие
истинно, - им пользуются, когда заранее неизвестно, сколько шагов понадобится.

````python
i = 0
while i < 3:
    print(i, LABELS[i])
    i += 1
````

> **Строка `i += 1` обязательна.** Забудете - условие никогда не станет
> ложным, и цикл не закончится. Блок кода зависнет.
>
> Поэтому правило: **`for` везде, где число шагов известно заранее.**
> `while` - только там, где оно действительно неизвестно: ждём ввода
> пользователя, читаем поток до стоп-значения, повторяем запрос до успеха.

# 6. Словарь

Список хранит значения по номерам. Но номер редко осмыслен: Вам нужно
«сколько записей в рубрике *наука*», а не «что лежит в ячейке 4».

Словарь хранит значения **по ключу** - по слову, числу или кортежу.
Ключи уникальны.

````python
counts = {"наука": 2, "спорт": 1}

print(counts)
print(counts["наука"])
print(len(counts))

counts["экономика"] = 4
counts["спорт"] = 2
print(counts)
````

Присваивание по ключу работает в обе стороны: ключ есть - значение
заменится, ключа нет - пара добавится. А вот **чтение** несуществующего
ключа - вызовет ошибку `KeyError`.

Два способа не получать ошибки с помощью встроенных методов:

- **`ключ in словарь`** - проверить заранее;
- **`.get(ключ, умолчание)`** - вернуть значение, а если ключа нет -
  умолчание. Словарь при этом не меняется.

`.get()` без второго аргумента вернёт `None` - «ничего нет». Это не ноль
и не пустая строка, складывать с ним нельзя, поэтому умолчание почти всегда
указывают явно.

````python
counts = {"наука": 2, "спорт": 1}

print("наука" in counts)
print(counts.get("погода"))
print(counts.get("погода", 0))
print(counts)

# раскомментируйте:
# print(counts["погода"])
````

## Цикл по словарю

`for ключ in словарь` идёт по ключам. Чаще нужны оба - для этого используйте **`.items()`**:
он выдаёт пары и раскладывает их по двум именам, как `zip`.

````python
counts = {"наука": 2, "спорт": 1, "экономика": 4}

for label, number in counts.items():
    print(label.ljust(12), number)

print("всего:", sum(counts.values()))
print("по алфавиту:", dict(sorted(counts.items())))
````

# 7. Частотный словарь

Главный приём всего курса. Задача: посчитать, сколько раз встречается каждое
значение, когда заранее неизвестно, какие значения будут.

Часто, чтобы посчитать, сколько раз встречается каждое значение в тексте, когда заранее оно неизвестно, создают словарь наполняют одной строкой:

```python
freq[x] = freq.get(x, 0) + 1
```

Читается так: «взять текущее значение **или ноль, если такого ещё не было**,
прибавить единицу, положить обратно». Одна строка делает и создание ключа,
и увеличение счётчика - ради этого и нужен был `.get()`.

Перед подсчётом слов текст приводят к нижнему регистру и убирают знаки
препинания, иначе `Учёные`, `учёные` и `учёные,` окажутся тремя разными словами.

````python
all_text = " ".join(TEXTS).lower()
for sign in ".,!?:;":
    all_text = all_text.replace(sign, " ")

words = all_text.split()

freq = {}
for word in words:
    freq[word] = freq.get(word, 0) + 1

print("словоупотреблений:", len(words))
print("уникальных слов:", len(freq))
print("в     :", freq.get("в", 0))
print("новый :", freq.get("новый", 0))
print("учёные:", freq.get("учёные", 0))
print("ученые:", freq.get("ученые", 0))
````

> Последние две строки - не опечатка в данных, а настоящая проблема.
> «Учёные» и «учёными» словарь тоже считает разными словами, хотя это одно
> и то же слово в разных формах. Приведение слова к начальной форме
> называется **лемматизацией**, и это тема отдельного занятия.
>
> А пара «ё» / «е» обойдётся вам ровно в половину частоты каждого такого
> слова - и, в отличие от лемматизации, чинится одной заменой.

## Тот же приём считает что угодно

Обратите внимание: чтобы посчитать не слова, а метки, меняется **одно слово**.
Ни одна строка внутри цикла не меняется - счётчику всё равно, что считать.

````python
freq = {}
for label in LABELS:
    freq[label] = freq.get(label, 0) + 1
print(freq)

freq = {}
for score in SCORES:
    freq[score] = freq.get(score, 0) + 1
print(freq)
````

# 8. Сортировка по частоте

Словарь напрямую не сортируется - сортируется список. Приём: переложить пары
в список кортежей **`(число, имя)`** и отсортировать.

Почему в таком порядке? Кортежи сравниваются по первому элементу, при
равенстве - по второму. Поставив число первым, вы получаете сортировку
по числу без всяких дополнительных параметров.

````python
freq = {"в": 3, "новый": 2, "учёные": 1, "вид": 2}

pairs = []
for word, count in freq.items():
    pairs.append((count, word))

print(pairs)
pairs.sort(reverse=True)
print(pairs)

for count, word in pairs[:3]:
    print(word.ljust(10) + str(count).rjust(3) + "  " + "*" * count)
````

> **Тот же результат можно получить короче** - `sorted(freq.items(), key=lambda pair: -pair[1])`.
> Запись с `key` и `lambda` вы будете встречать в чужом коде постоянно,
> и в теме 2 разберёте её как следует. Пока достаточно приёма с кортежами:
> он делает ровно то же самое и не требует ничего, кроме уже известного.

````python
groups = {}
for label, score in zip(LABELS, SCORES):
    if label not in groups:
        groups[label] = []
    groups[label].append(score)

print(groups)

for label, scores in groups.items():
    print(f"{label.ljust(8)} n={len(scores)} {scores} среднее={sum(scores) / len(scores):.2f}")
````

> Рядом со средним **всегда** находится `n`. Среднее по одной записи -
> это не среднее, а просто запись, и выглядит оно в таблице точно так же,
> как честное среднее по пятистам. Это не придирка к оформлению,
> а способ не сделать неверный вывод.
>
> У словарей есть готовый метод для этого приёма:
> `groups.setdefault(ключ, []).append(значение)` — одна строка вместо трёх.
> Делает он ровно то же самое.

# 9. `match` / `case`

Когда вариантов много, цепочка `if / elif / elif / else` становится длинной,
и в каждой строке повторяется одно и то же имя. Для таких случаев существует `match`
(появился в Python 3.10).

Значение сверху по очереди сравнивается с образцами, срабатывает **первый
подходящий**. `case _` - «всё остальное», пишется последним. Несколько
вариантов в одной ветке перечисляют через `|`.

````python
for score in [5, 4, 3, 2, 1, 99]:
    match score:
        case 5:
            print(score, "-> отлично")
        case 4:
            print(score, "-> хорошо")
        case 3:
            print(score, "-> средне")
        case 1 | 2:
            print(score, "-> плохо")
        case _:
            print(score, "-> неизвестная оценка")
````

> **Главная ловушка `match`.** `case score:` без кавычек - это **не**
> сравнение с переменной `score`, а создание новой переменной с таким именем.
>
> Для сравнения пишите само значение: `case 5:`, `case "наука":`.
> Ошибки не будет - будет ветка, которая срабатывает на всём подряд.

# 11. `input()` - ввод с клавиатуры

`input()` показывает приглашение, ждёт, пока человек напечатает строку
и нажмёт Enter, и возвращает её.

Две вещи, которые надо помнить:

- **`input()` всегда возвращает строку**, даже если введено число.
  Чтобы получить число, необходимо преобразовать строку в целочисленную переменную пишут `int(input())`.
- в скобках `input()`, можно писать поясняющий комментарий для ввода от пользователя - `input("Комментарий для пользователя")`

Типичная конструкция - `while True` с `break` внутри: бесконечный цикл,
из которого выходят по команде.

````python
name = input("Введите свое имя: ")
print("Привет, ", name)
age = int(input("Введите Ваш возраст: "))
print(f"Ваш возраст - {age} лет")
````

````python
# Здесь ввод не запрашивается — команды взяты из списка,
# чтобы конспект можно было запускать целиком.
# В практике вы замените COMMANDS на while True + input().

COMMANDS = ["stats", "show 1", "привет", "quit"]

for line in COMMANDS:
    print(">", line)
    parts = line.split()
    command = parts[0]

    match command:
        case "stats":
            print("записей:", len(TEXTS))
        case "show":
            number = int(parts[1])
            print(LABELS[number], "|", TEXTS[number][:30])
        case "quit":
            print("До встречи.")
            break
        case _:
            print("Команды: stats | show N | quit")
````

> Обратите внимание на `parts[1]`: если человек напечатал просто `show`,
> без номера, этой части не существует и программа упадёт с `IndexError`.
> Поэтому перед обращением проверяют `len(parts) > 1`.
>
> Это общее правило работы с вводом: **всё, что ввёл человек, может быть
> не тем, что вы ждёте.** Пустая строка, буквы вместо числа, номер вне
> диапазона - проверять нужно каждое.

# Шпаргалка занятия

| Что нужно | Как пишется | Замечание |
| --- | --- | --- |
| сделать с каждым | `for x in список:` | |
| тело цикла | отступ в 4 пробела | это синтаксис, а не оформление |
| по символам | `for ch in строка:` | |
| по номерам | `for i in range(n):` | `range(a, b, шаг)` |
| накопить сумму | `total = 0` **до** цикла, `total += x` внутри | |
| накопить список | `res = []` до цикла, `res.append(x)` внутри | |
| номер и элемент | `for i, x in enumerate(список, start=1):` | |
| несколько списков | `for a, b in zip(A, B):` | останавливается на коротком |
| номер и пара | `for i, (a, b) in enumerate(zip(A, B)):` | скобки обязательны |
| если / иначе если / иначе | `if:` / `elif:` / `else:` | срабатывает первое подходящее |
| сравнение | `==`, `!=`, `>`, `<`, `>=`, `<=` | `==`, не `=` |
| и / или / не / входит | `and`, `or`, `not`, `in` | |
| проверить `bool` | `if verified:` | без `== True` |
| пропустить шаг | `continue` | |
| выйти из цикла | `break` | |
| пока верно | `while условие:` | не забыть менять условие |
| словарь | `{"ключ": значение}` | ключи уникальны |
| по ключу | `d["ключ"]` | `KeyError`, если ключа нет |
| безопасно | `d.get("ключ", 0)` | словарь не меняется |
| есть ли ключ | `"ключ" in d` | |
| цикл по парам | `for k, v in d.items():` | |
| сумма значений | `sum(d.values())` | |
| **частотный словарь** | `d[x] = d.get(x, 0) + 1` | главный приём |
| **словарь списков** | `if k not in d: d[k] = []`, затем `d[k].append(v)` | группировка |
| ключ из двух полей | `d[(a, b)] = d.get((a, b), 0) + 1` | кортеж можно, список нельзя |
| сортировка по числу | список кортежей `(число, имя)` + `.sort(reverse=True)` | |
| по алфавиту ключей | `dict(sorted(d.items()))` | |
| много вариантов | `match x:` / `case 1 \| 2:` / `case _:` | `case имя:` — ловушка |
| ввод | `input()` | всегда строка, нужен `int()` |

Полный перечень методов - в «Справочнике по основам Python».

# Частые ошибки второго занятия

**1. Нет отступа в теле цикла.** `IndentationError`. Python не догадывается,
что строку надо было положить внутрь.

**2. Счётчик объявлен внутри цикла.** `total = 0` на каждом шаге обнуляется,
в конце остаётся последнее значение. **Ошибки нет - есть неверное число**,
и это самая дорогая ошибка из всех перечисленных.

**3. Изменение переменной цикла не меняет список.**
`for w in words: w = w.upper()` не делает ничего: `w` - это имя для копии
значения. Чтобы изменить список, собирают новый через `.append()`
или пишут `words[i] = ...`.

**4. Список меняется во время прохода по нему.** Удаление элементов внутри
`for` по этому же списку пропускает элементы. Собирайте новый список.

**5. `KeyError`.** Читаете ключ, которого нет. Нужен `.get(ключ, 0)`.

**6. Забыт `.lower()` перед поиском.** Слово есть, а `in` говорит `False`.

**7. Бесконечный `while`.** Условие не меняется внутри цикла.
Останавливать кнопкой «стоп» (■).

**8. `case имя:` вместо `case "значение":`.** Ветка срабатывает всегда,
остальные не проверяются. Ошибки нет.

**9. Порядок веток в `if / elif`.** Более широкое условие стоит выше более
узкого, и узкое не срабатывает никогда.

**10. `input()` вернул строку.** `input() + 1` - `TypeError`,
`input() > 5` - тоже. Нужен `int()`.
