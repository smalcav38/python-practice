## Что вы сможете к концу занятия

Вот программа, которую вы соберёте в практике. Каждая строка внизу - вызов
функции, которую Вы напишете сами.

````python
def clean(text):
    """Приводит к нижнему регистру и убирает знаки препинания."""
    text = text.lower()
    for sign in ".,!?:;":
        text = text.replace(sign, " ")
    return text

def words(text):
    """Список слов текста."""
    return clean(text).split()

def word_count(text):
    return len(words(text))

texts = ["Кот спит.", "Собака бежит по улице!", "Кот ест, потом спит."]
for text in texts:
    print(word_count(text), "|", text)
````

# 1. Зачем нужны функции

Посмотрите на код, здесь нужно посчитать слова
в каждой записи и сказать, длинная она или короткая.

````python
count = len(TEXTS[0].split())
print(count, "длинная" if count > 7 else "короткая")

count = len(TEXTS[1].split())
print(count, "длинная" if count > 7 else "короткая")

count = len(TEXTS[2].split())
print(count, "длинная" if count > 7 else "короткая")
````

Работает. А теперь Вы решили, что «длинная» - это больше **шести** слов,
а не семи. Исправлять придётся в трёх местах. Забудете одно и программа

Это одна из важнейших причин исаользовать функции.
Не «чтобы было короче» - а чтобы одно решение было записано **один раз**.

# 2. `def` и вызов

Функцию создают при помощи ключевого слова `def`: имя, круглые скобки, двоеточие. Тело —
с отступом, как у цикла.

```python
def имя(параметры):
    тело
```

**Определение ничего не выполняет.** Строка `def` только запоминает код
под именем. Выполнится он, когда функцию **вызовут**: имя и круглые скобки.

````python
def say_hello():
    print("Привет!")
    print("Я — функция.")

print("функция определена, но ещё ничего не напечатала")

say_hello()
say_hello()
````

Две вещи, на которых ошибаются:

- **Без скобок функция не вызывается.** `say_hello` - это сама функция как объект,
  `say_hello()` - её вызов. Точно так же, как с методами в теме 1.
- **Вызвать можно только то, что уже определено.** Если ячейку с `def`
  не запускали, выведется ошибка `NameError` - Python ещё не знает такого имени.

````python
print(say_hello)     # описание функции, а не вызов

# раскомментируйте — такой функции ещё нет:
# say_goodbye()
````

# 3. Параметры и аргументы

Функция без параметров всегда делает одно и то же. Параметр - это переменная
в скобках, которую функция получает снаружи.

Два понятия, которые важно различать:

- **параметр** - имя в определении: `def greet(name):` - здесь `name`;
- **аргумент** - значение при вызове: `greet("Анна")` - здесь `"Анна"`.

При вызове аргумент присваивается в параметр, и дальше внутри функции
с ним работают как с обычной переменной.

````python
def greet(name):
    print(f"Привет, {name}!")

greet("Анна")
greet("Борис")

def show_record(label, score):
    print(f"{label}: оценка {score}")

show_record("наука", 5)
show_record(LABELS[1], SCORES[1])
````

Аргументы подставляются **по порядку**: первый - в первый параметр,
второй - во второй. Перепутали порядок и функция получила не то.
Ошибки при этом не будет.

````python
show_record(5, "наука")   # порядок перепутан — результат бессмысленный

# раскомментируйте — аргументов меньше, чем параметров:
# show_record("наука")
````

# 4. `return` против `print`

**Это главный раздел конспекта.** Самая частая ошибка с функциями выглядит так:
функция всё правильно считает, печатает ответ и этим ответом нельзя
воспользоваться.

- **`print`** показывает значение **человеку**. Дальше оно пропадает.
- **`return`** отдаёт значение **программе**. Его можно присвоить переменной,
  сложить, сравнить, передать другой функции.

````python
def count_words_print(text):
    print(len(text.split()))

def count_words(text):
    return len(text.split())

a = count_words_print(TEXTS[0])
b = count_words(TEXTS[0])

print("a =", a)
print("b =", b)
````

Обе функции «посчитали» 9 слов. Но первая только **показала** девятку,
а вернула пустоту - `None`. Вторая вернула число, и с ним можно работать дальше.

````python
total = 0
for text in TEXTS:
    total += count_words(text)
print("слов во всех записях:", total)

print("длиннее ли первая третьей:", count_words(TEXTS[0]) > count_words(TEXTS[2]))

# раскомментируйте - с None складывать нельзя:
# print(count_words_print(TEXTS[0]) + 1)
````

> **Правило:** функция, которая что-то **считает**, возвращает результат
> через `return`. Печатает — тот, кто её вызвал. Функции, которые только печатают,
> тоже бывают — но это функции-отчёты, и их всегда видно по имени:
> `print_report`, `show_table`.

# 5. `None` - «ничего»

Если функция не дошла до `return` (или его нет вовсе), она возвращает
особое значение **`None`**. Это не ноль, не пустая строка и не `False` -
это отдельное «ничего».

`None` вы уже видели в теме 1: его возвращает `.sort()`. Теперь понятно почему:
`.sort()` меняет список и ничего не возвращает.

````python
def nothing():
    x = 5

result = nothing()
print(result)
print(type(result))

print([3, 1, 2].sort())
````

# 6. `return` завершает функцию

Как только выполнился `return`, функция **сразу заканчивает свое выполнение**. Строки ниже
не выполняются. Этим пользуются, чтобы выйти раньше, например, когда
данные не подходят.

````python
def average_length(texts):
    if len(texts) == 0:
        return 0
    total = 0
    for text in texts:
        total += len(text.split())
    return total / len(texts)

print(average_length(TEXTS))
print(average_length([]))    # без первого return было бы деление на ноль
````

## Функция, которая отвечает «да» или «нет»

Функция может вернуть `True` или `False`. Такие функции называют
**предикатами**, и им дают имена, которые читаются как вопрос:
`is_long`, `has_digits`, `is_positive`. Их удобно ставить прямо в `if`.

````python
def is_long(text):
    return len(text.split()) > 7

for text in TEXTS:
    if is_long(text):
        print("длинная:", text[:30])

print(is_long(TEXTS[1]))
````

> Обратите внимание: не `if ...: return True else: return False`,
> а сразу `return условие`. Условие уже **есть** `True` или `False`.

Через запятую после `return` можно вернуть несколько значений. На самом деле

## Словарь как результат

Если значений больше двух-трёх, вызывающему трудно помнить их порядок.
Тогда возвращают **словарь**: у каждого числа появляется имя.

````python
def text_stats(text):
    return {
        "chars": len(text),
        "words": len(text.split()),
        "unique": len(set(text.lower().split())),
    }

stats = text_stats(TEXTS[0])
print(stats)
print(stats["words"])
````

# 8. Значения по умолчанию

Параметру можно задать значение, которое подставится, если аргумент
не передали. Так делают для «классического» случая: чаще всего нужно одно,
иногда - другое.

Параметры со значением по умолчанию пишутся **после** обязательных.

````python
def shorten(text, width=20):
    if len(text) <= width:
        return text
    return text[:width] + "..."

print(shorten(TEXTS[0]))        # width взялось по умолчанию
print(shorten(TEXTS[0], 40))
print(shorten(TEXTS[1], 100))
````

## Именованные аргументы

При вызове можно указать **имя** параметра: `shorten(text, width=40)`.
Тогда порядок не важен, а вызов читается без заглядывания в определение:
`shorten(t, 40)` - непонятно, что такое 40; `shorten(t, width=40)` - понятно.

Позиционные аргументы всегда пишутся **перед** именованными.

````python
print(shorten(TEXTS[2], width=15))
print(shorten(width=15, text=TEXTS[2]))

# раскомментируйте - позиционный после именованного:
# print(shorten(width=15, TEXTS[2]))
````

# 9. Docstring и `help()`

Первая строка в теле функции, записанная в тройных кавычках, называется
**строкой документации** - docstring. Python её не выполняет, а сохраняет:
её показывает `help()`, и её же показывают подсказки в Colab и VS Code.

Хорошая строка документации отвечает на вопрос «что вернёт функция»,
а не «как она устроена».

````python
def share(values, threshold):
    """Доля значений, которые не меньше threshold. Для пустого списка - 0."""
    if len(values) == 0:
        return 0
    count = 0
    for value in values:
        if value >= threshold:
            count += 1
    return count / len(values)

print(share(SCORES, 4))
help(share)
````

# 10. Функция вызывает функцию

Функции можно собирать друг из друга. Маленькая функция делает одно дело,
большая - вызывает несколько маленьких.

Это и есть главный способ писать программы: не одна ячейка на сто строк,
а десять функций по пять строк, каждую из которых можно проверить отдельно.

````python
def clean(text):
    """Нижний регистр, знаки препинания заменены пробелами."""
    text = text.lower()
    for sign in ".,!?:;":
        text = text.replace(sign, " ")
    return text

def words(text):
    """Список слов текста после очистки."""
    return clean(text).split()

def frequencies(items):
    """Частотный словарь: {элемент: сколько раз}."""
    freq = {}
    for item in items:
        freq[item] = freq.get(item, 0) + 1
    return freq

print(words(TEXTS[1]))
print(frequencies(LABELS))

all_words = []
for text in TEXTS:
    all_words += words(text)
print(frequencies(all_words)["новый"])
````

Обратите внимание на `frequencies`. Ей всё равно, что считать: слова, метки,
оценки.

А ещё взгляните на последние строки: `3:1` из второй записи стало двумя
«словами» - `3` и `1`. Функция `clean` приняла это решение за вас. Когда
вы захотите решить иначе, исправлять придётся **только её** и всё,
что вызывает `words`, изменится само.

# 11. Область видимости

Переменные, созданные **внутри** функции называются **локальными**. Они появляются при
вызове и исчезают, когда функция закончилась. Вне функции у ним обратиться нельзя.

````python
def compute():
    inner = 42
    return inner * 2

print(compute())

# раскомментируйте — inner снаружи не существует:
# print(inner)
````

Переменные, созданные **снаружи** всех функций называются **глобальными**.
Читать их внутри функции можно, но тогда функция зависит от того,
что находится в переменной снаружи, и перестаёт правильно работать, если ее переименовали.

Поэтому правило: **всё, что нужно функции, она получает через параметры,
а всё, что посчитала, возвращает через `return`.**

````python
LIMIT = 7

def is_long_global(text):
    return len(text.split()) > LIMIT        # зависит от того, что снаружи

def is_long_param(text, limit=7):
    return len(text.split()) > limit        # всё нужное - в параметрах

print(is_long_global(TEXTS[0]), is_long_param(TEXTS[0]), is_long_param(TEXTS[0], limit=10))
````

## `UnboundLocalError`

Если внутри функции выполняется **присваивание** какой-то переменной, Python считает
это имя локальным **во всей функции** - даже в строках выше присваивания.
Попытка сначала прочитать глобальную переменную, а потом изменить её,
приведет к ошибке `UnboundLocalError`.

````python
total = 0

def add_words(text):
    total = total + len(text.split())   # total здесь уже локальная — а её ещё нет
    return total

# раскомментируйте и прочитайте последнюю строку ошибки:
# add_words(TEXTS[0])
````

Существует ключевое слово `global`, которое разрешает такое изменение. Пользоваться им нужно с осторожностью:
функция, которая изменяет что-то снаружи, непредсказуема, и искать, кто
изменил переменную, приходится долго. Вместо этого используйте параметр и `return`:

````python
def add_words(total, text):
    return total + len(text.split())

total = 0
for text in TEXTS:
    total = add_words(total, text)
print(total)
````

# 12. Что происходит с аргументом

Когда вы передаёте в функцию список, **копии не создаётся**. Параметр внутри
функции - это просто ещё одно имя для **того же самого** списка:

- функция **изменяет объект** - `.append()`, `.remove()`, `d[k] = v`: изменение
  видно снаружи, ведь объект один;
- функция **присваивает параметру новое значение** - `x = ...`: внутри
  имя начинает указывать на новый объект, снаружи ничего не изменяется.

````python
def spoil(values):
    values.append(999)        # изменяет тот же список

def rebind(values):
    values = values + [999]   # создаёт новый, исходный не изменяет

data = [1, 2, 3]
spoil(data)
print("после spoil :", data)

data = [1, 2, 3]
rebind(data)
print("после rebind:", data)
````

А что с числами и строками? Их изменить **нельзя** в принципе.
Любая операция создаёт новый объект, то есть это всегда второй
случай. Поэтому кажется, будто «числа в функции не изменяются, а списки изменяются».
На самом деле разница не в типе, а в том, **что делает функция**.

````python
def add_ten(number):
    number += 10              # новое число, старое не тронуто
    return number

x = 5
add_ten(x)
print(x)

y = add_ten(x)                # результат забирают через return
print(y)
````

> **Правило, которое сэкономит вам дни отладки.** Функция не изменяет то,
> что ей передали. Если нужно получить изменённые данные, функция создаёт
> **новый** список и возвращает его через `return`. Исходный остаётся без изменений.

````python
def without_short(words, min_len=3):
    """Новый список без коротких слов. Исходный не меняется."""
    result = []
    for word in words:
        if len(word) >= min_len:
            result.append(word)
    return result

words = TEXTS[1].split()
print(without_short(words))
print(words)
````

# 13. Ловушка: список по умолчанию

Значение по умолчанию вычисляется **один раз** в момент, когда выполняется
строка `def`. Если это список, все вызовы получают **один и тот же** список.
И он копит данные между вызовами.

````python
def collect(item, bag=[]):      # ЛОВУШКА
    bag.append(item)
    return bag

print(collect("а"))
print(collect("б"))             # список из прошлого вызова никуда не делся
````

В эту ловушку попадают все, и один раз. Правильная запись использование `None`
по умолчанию и новый список внутри функции:

````python
def collect(item, bag=None):
    if bag is None:
        bag = []
    bag.append(item)
    return bag

print(collect("а"))
print(collect("б"))
````

> `is None`  так принято проверять именно `None`. Для чисел и строк
> по-прежнему пишут `==`.

# 14. `*args`: сколько угодно аргументов

Звёздочка перед параметром говорит: «собери все лишние позиционные аргументы
в **кортеж**». Сколько передали, столько и попадёт.

````python
def total_words(*texts):
    print("  texts =", texts)
    total = 0
    for text in texts:
        total += len(text.split())
    return total

print(total_words(TEXTS[0]))
print(total_words(TEXTS[0], TEXTS[1], TEXTS[2]))
print(total_words())
````

Звёздочка работает и в обратную сторону при **вызове**: `f(*список)`
раскладывает список на отдельные аргументы.

````python
print(total_words(*TEXTS))    # то же, что total_words(TEXTS[0], TEXTS[1], TEXTS[2])
````

# 15. `**kwargs`: сколько угодно именованных

Две звёздочки собирают все лишние **именованные** аргументы в **словарь**:
имя аргумента становится ключом, значение - значением.

````python
def describe_columns(**columns):
    for name, values in columns.items():
        print(name.ljust(8), "n =", len(values), " сумма =", sum(values))

describe_columns(scores=SCORES, lengths=[len(t.split()) for t in TEXTS])
````

> Имена `args` и `kwargs` - соглашение, а не требование. Важны звёздочки.
> Хорошие имена лучше: `*texts`, `**columns` сразу говорят, что внутри.

# 16. Функция - это значение

Имя функции **без скобок** такое же значение, как число или строка.
Его можно присвоить другой переменной, в список, в словарь
и **передать в другую функцию**.

Это звучит как головоломка, но вы пользовались этим с первого занятия:
`sorted(words, key=len)` это передача функции `len` в функцию `sorted`.

````python
def word_count(text):
    return len(text.split())

def char_count(text):
    return len(text)

counter = word_count          # без скобок: не результат, а сама функция
print(counter(TEXTS[0]))

for func in [word_count, char_count]:
    print(func.__name__, func(TEXTS[0]))
````

## Функция, которая принимает функцию

Если функция принимает другую функцию параметром, одну и ту же логику можно
применять с разными правилами не переписывая её.

````python
def apply_to_all(func, items):
    """Новый список: func применена к каждому элементу."""
    result = []
    for item in items:
        result.append(func(item))
    return result

print(apply_to_all(word_count, TEXTS))
print(apply_to_all(char_count, TEXTS))
print(apply_to_all(str.upper, LABELS))
````

# 17. `key=` и `lambda`

Главное место, где функцию передают в функцию, параметр **`key`** у `sorted`,
`min` и `max`. `key` - это функция, которую применяют к каждому элементу,
**прежде чем сравнивать**. Сравниваются уже её результаты.

````python
print(sorted(TEXTS, key=word_count))
print(max(TEXTS, key=word_count))

counts = [("новость", 5), ("учёные", 12), ("вирус", 7)]

def by_count(pair):
    return pair[1]

print(sorted(counts, key=by_count, reverse=True))
````

Писать отдельную функцию `by_count` ради одной строки неудобно. Для таких
случаев есть **`lambda`** - функция из одного выражения, записанная прямо там,
где нужна:

```python
lambda pair: pair[1]        # то же, что def by_count(pair): return pair[1]
```

Слева от двоеточия - параметры, справа - выражение, которое вернётся.
`return` не пишется.

````python
print(sorted(counts, key=lambda pair: pair[1], reverse=True))
print(min(counts, key=lambda pair: pair[1]))
````

## Ключ из двух частей

Часто нужно сортировать сначала по одному признаку, а **при равенстве** -
по другому: по частоте вниз, при равной частоте - по алфавиту. Ключ тогда
возвращает **кортеж**: кортежи сравниваются по первому элементу, при равенстве -
по второму.

А чтобы отсортировать наоборот, число пишут с минусом.

````python
pairs = [("в", 3), ("а", 3), ("б", 5)]

print(sorted(pairs, key=lambda p: -p[1]))
print(sorted(pairs, key=lambda p: (-p[1], p[0])))
````

> В первой строке «в» и «а» с одинаковой частотой остались в том же порядке,
> случайном по смыслу. Во второй порядок задан правилом.
> Функция, чей результат зависит от случайного порядка входа, не поддаётся
> проверке: сегодня один ответ, завтра другой.
>
> **Когда использовать `lambda`, а когда `def`?** Лямбда - если выражение короткое и нужно
> в одном месте. Если ключ не помещается в строку или нужен дважды -
> `def` с понятным именем.

# 18. Списковое включение, `map`, `filter`

Цикл, который строит новый список, можно записать одной строкой:

```python
[выражение for элемент in последовательность if условие]
```

Читается слева направо: «выражение - для каждого элемента - если условие».
Часть с `if` необязательна.

````python
lengths = []
for text in TEXTS:
    lengths.append(len(text.split()))
print(lengths)

print([len(text.split()) for text in TEXTS])
print([text for text, label in zip(TEXTS, LABELS) if label == "наука"])
print([word.lower() for word in TEXTS[1].split() if word.isalpha()])
````

То же самое умеют встроенные `map` (применить функцию к каждому) и `filter`
(оставить подходящие). Они возвращают «ленивый» результат и чтобы увидеть его,
оборачивают в `list()`. Включение обычно читается проще, но `map` и `filter`
вы будете часто встречать в чужом коде.

````python
print(list(map(word_count, TEXTS)))
print(list(filter(lambda t: word_count(t) > 7, TEXTS)))
````

## Круглые скобки вместо квадратных

Если включение записать в круглых скобках, получится **генератор** он выдаёт
значения по одному и **читается только один раз**. Подробно рассматривать генераторы будем в следующем занятии.
Сейчас важно одно следствие: функция, которая принимает «что угодно
перебираемое» должна один раз сложить его в список и дальше работать со списком.

````python
lengths = (len(t.split()) for t in TEXTS)   # генератор
print(sum(lengths))
print(sum(lengths))                          # пусто: уже прочитан

def mean(values):
    values = list(values)                    # складываем один раз
    if len(values) == 0:
        return 0
    return sum(values) / len(values)

print(mean(len(t.split()) for t in TEXTS))
````

# 19. Функция, которая возвращает функцию

Функцию можно определить **внутри** другой функции и вернуть наружу.
Так собирают «рецепты»: описание, что и в каком порядке сделать,
без самого выполнения.

Внутренняя функция помнит параметры внешней, даже после того, как внешняя
функция закончила свое выполнение.

````python
def longer_than(limit):
    def check(text):
        return word_count(text) > limit
    return check

is_long = longer_than(7)
is_very_long = longer_than(8)

print(is_long(TEXTS[0]), is_very_long(TEXTS[0]))
````

Тот же приём с `*args` даёт **конвейер**: функцию, которая применяет
несколько функций по очереди.

````python
def chain(*funcs):
    def apply(value):
        for func in funcs:
            value = func(value)
        return value
    return apply

def add_one(x):
    return x + 1

def double(x):
    return x * 2

print(chain(add_one, double)(5))   # (5 + 1) * 2
print(chain(double, add_one)(5))   # 5 * 2 + 1 — порядок важен

prepare = chain(str.lower, str.split)
print(prepare(TEXTS[1]))
````

# 20. Аннотации типов

После параметра через двоеточие можно указать, какого типа он ожидается,
а после скобок через `->` - что функция вернёт.

```python
def word_count(text: str) -> int:
```

**Python аннотации не проверяет.** Передадите число вместо строки, функция
запустится и закончит свое выполнение на первой же операции со строкой. Аннотации пишут для
людей и для редакторов кода: VS Code и Colab по ним подсказывают и заранее
подчёркивают подозрительные места.

````python
def word_count(text: str) -> int:
    return len(text.split())

def top(freq: dict[str, int], n: int = 3) -> list[tuple[str, int]]:
    return sorted(freq.items(), key=lambda p: (-p[1], p[0]))[:n]

print(word_count(TEXTS[0]))
print(top({"а": 1, "б": 3, "в": 2}))
print(word_count.__annotations__)
````

> Читается это так: `list[tuple[str, int]]` - «список кортежей, в каждом
> строка и целое число». Именно такие записи вы увидите в документации
> библиотек и в заготовках заданий.

# 21. Когда функции передали не то что она ожидает

Иногда функция получает то, с чем работать не может: число вместо строки,
`None` вместо списка. 

Стоит **сразу остановиться и сказать, что не так**. Для этого используют:

- **`isinstance(значение, тип)`** - проверить тип: `isinstance("кот", str)` -> `True`;
- **`raise TypeError("что не так")`** - остановить программу с ошибкой
  и понятным сообщением.

````python
def word_count(text):
    if not isinstance(text, str):
        raise TypeError(f"нужна строка, а передано {type(text).__name__}")
    return len(text.split())

print(word_count(TEXTS[0]))
print(isinstance(42, str), isinstance("42", str))

# раскомментируйте и прочитайте последнюю строку:
# word_count(42)
````

# 22. Как проверить функцию

Функцию проверяют так: вызывают с данными, для которых ответ **известен
заранее**, и сравнивают.

````python
def check(name, got, expected):
    if got == expected:
        print(f"  ok   {name}")
    else:
        print(f"  FAIL {name}: получено {got!r}, ожидалось {expected!r}")

check("три слова", word_count("кот спит тут"), 3)
check("пустая строка", word_count(""), 0)
check("лишние пробелы", word_count("  кот   спит "), 2)
check("специально неверная", word_count("кот"), 2)
````

Хороший набор проверок включает не только «обычные» случаи, но и **критичные**:
пустую строку, один элемент, одинаковые значения, неподходящий тип, именно там чаще возникают ошибки.

В больших проектах такие проверки хранят в отдельных файлах - это **тесты**,
и запускает их команда `pytest`. Устроены они ровно так же, как `check` выше:
имя проверки, что получилось, что ожидалось. Если тест не проходит - **читайте
его имя**: оно говорит, какое именно требование не выполнено.

## Одно полезное свойство: идемпотентность

Функцию называют **идемпотентной**, если повторное применение ничего не меняет:
`f(f(x)) == f(x)`. Функции очистки текста обязаны этим свойством обладать
иначе текст, случайно очищенный дважды, будет отличаться от очищенного однажды.

````python
def squash(text):
    return " ".join(text.split())      # убирает лишние пробелы

def cut(text):
    return text[:-1]                    # отрезает последний символ

value = "  кот   спит  "
check("squash идемпотентна", squash(squash(value)), squash(value))
check("cut идемпотентна", cut(cut(value)), cut(value))
````

## `main()`

Когда функций становится много, их собирают в одну - обычно её называют `main`.
Внутри нет вычислений, только вызовы. Вся программа становится одной строкой,
а каждый шаг можно проверить отдельно.

````python
def load():
    return TEXTS

def lengths(texts):
    return [word_count(t) for t in texts]

def report(values):
    print("  записей:", len(values), " слов:", sum(values), " в среднем:", round(mean(values), 2))

def main():
    report(lengths(load()))

main()
````

> В файлах `.py` вы увидите строку `if __name__ == "__main__": main()`.
> Она означает «вызвать `main()`, только если файл запустили напрямую, а не
> импортировали из другого файла». Импорт - **тема 6**; пока достаточно узнавать
> эту строку в чужом коде.

# Шпаргалка занятия

**Основы**

| Что нужно | Как пишется | Замечание |
| --- | --- | --- |
| определить функцию | `def имя(параметры):` | тело с отступом |
| вызвать | `имя(аргументы)` | без скобок - не вызов |
| вернуть результат | `return значение` | функция сразу заканчивается |
| ничего не вернуть | нет `return` | вернётся `None` |
| вернуть несколько | `return a, b` | это кортеж, распаковать `x, y = f()` |
| вернуть много | `return {"n": ..., "mean": ...}` | у значений есть имена |
| значение по умолчанию | `def f(x, width=20):` | после обязательных |
| именованный аргумент | `f(text, width=40)` | после позиционных |
| документация | `"""Что возвращает."""` первой строкой | `help(f)` |
| функция-предикат | `def is_long(t): return len(t) > 7` | имя как вопрос |
| функция из функций | `def words(t): return clean(t).split()` | одно дело — одна функция |
| локальная переменная | создана внутри функции | снаружи не видна |
| глобальную не менять | параметр + `return` | вместо `global` |

**Гибкость**

| Что нужно | Как пишется | Замечание |
| --- | --- | --- |
| функция меняет список | `values.append(...)` внутри | видно снаружи |
| функция не портит вход | создать новый, `return` его | правило курса |
| список по умолчанию | `def f(x, bag=None):` + `if bag is None: bag = []` | не `bag=[]` |
| сколько угодно позиционных | `def f(*items):` | внутри кортеж |
| сколько угодно именованных | `def f(**columns):` | внутри словарь |
| разложить список в аргументы | `f(*список)` | |
| функция как значение | `g = f` (без скобок) | |
| передать функцию | `apply(f, items)`, `sorted(x, key=f)` | |
| короткая функция на месте | `lambda p: p[1]` | одно выражение |
| сортировка по двум ключам | `key=lambda p: (-p[1], p[0])` | число вниз, слово вверх |
| список одной строкой | `[f(x) for x in xs if cond]` | |
| `map` / `filter` | `list(map(f, xs))`, `list(filter(f, xs))` | ленивые |
| генератор читается раз | `values = list(values)` в начале | |
| функция возвращает функцию | `def outer(): def inner(): ...; return inner` | |
| аннотации | `def f(text: str) -> int:` | Python не проверяет |
| проверить тип | `isinstance(x, str)` | |
| остановить с ошибкой | `raise TypeError("что не так")` | |
| проверить функцию | известный вход → известный ответ | и края тоже |

Полный перечень — в «Справочнике по функциям».
