## Особенности работы интерпретатора Python

Python является интерпретируемым языком программирования, что означает, что код выполняется интерпретатором строчка за строчкой. Это отличается от компилируемых языков, таких как C или Java, где код сначала компилируется в машинный код, а затем выполняется. Это делает Python очень удобным для быстрого прототипирования и интерактивной работы.

### Как работает интерпретатор Python?

- **REPL (Read-Eval-Print Loop):** Интерпретатор Python работает по принципу REPL. Это цикл, который:
  - Читает строку кода (Read).
  - Выполняет ее (Eval).
  - Выводит результат (Print).
  - Повторяет процесс (Loop).

- **Динамическая типизация:** В Python используется динамическая типизация, что означает, что тип переменной определяется во время выполнения программы, а не на этапе компиляции. Это упрощает процесс разработки, но требует осторожности при работе с типами данных.

- **Интерпретатор CPython:** Самая распространенная реализация Python — это CPython, написанная на языке C. Она обеспечивает хорошую совместимость с другими библиотеками, написанными на C, и поддерживает большинство популярных операционных систем.

### Переменные и типы данных

Переменная - это именованная область в памяти (хранилище данных), которая может изменяться в процессе выполнения программы.
Python имеет простую и интуитивно понятную систему работы с переменными. Переменная в Python создается в момент присвоения ей значения.

Динамическая типизация в Python означает, что тип данных переменной определяется автоматически во время выполнения программы, а не заранее, как в языках с статической типизацией (например, C или Java).

## Как это работает?
В Python, когда вы присваиваете значение переменной, интерпретатор автоматически определяет тип данных этой переменной на основе присвоенного значения. Например:

````python
x = 10       # x становится целым числом (int)
y = "Hello"  # y становится строкой (str)
z = 3.14     # z становится числом с плавающей запятой (float)
````

Здесь интерпретатор видит значение 10 и понимает, что это целое число, поэтому тип переменной x становится int. Аналогично, для y - str, для z - float.

Интерпретатор отслеживает тип переменной, и если вы присваиваете ей значение другого типа, тип переменной меняется:

````python
x = 10       # x сейчас int
x = "Hello"  # x теперь str
````

## Как интерпретатор определяет тип данных?
Python хранит тип данных вместе с объектом (значением), к которому ссылается переменная. Когда интерпретатор видит присваивание, он смотрит на объект и определяет его тип. Например:
- Целые числа (int): числа без дробной части, например, 1, 42.
- Строки (str): последовательности символов, например, "Hello".
- Списки (list): упорядоченные последовательности, например, [1, 2, 3].
- И так далее для других типов данных.

## Недостатки динамической типизации
- Ошибки во время выполнения: Поскольку типы данных не проверяются во время компиляции, вы можете столкнуться с ошибками уже во время выполнения программы.

````python
x = "10"
y = "20"
result = x + y  # Ошибка! Нельзя сложить int и str
print(result)
````

- **Трудности отладки**: Поскольку типы могут меняться динамически, отладка и понимание кода могут стать сложнее.
- **Производительность**: Из-за того, что Python должен отслеживать типы данных во время выполнения, программы могут работать медленнее по сравнению с языками со статической типизацией.

## Основные типы данных в Python:

- Числовые типы:
    - **int** (целые числа): Например, 5, -3, 0.
    - **float** (вещественные числа): Например, 5.0, -3.14, 0.001.
    - **complex** (комплексные числа): Например, 3+4j.
- Строки (**str**): Последовательность символов, например, "Hello, World!".
- Булевы значения (**bool**): Логические значения, которые могут быть либо True, либо False.
- Списки (**list**): Массивы значений, например, [1, 2, 3].
- Кортежи (**tuple**): Как списки, но неизменяемые, например, (1, 2, 3).
- Словари (**dict**): Набор пар "ключ-значение", например, {"ключ1": "значение1", "ключ2": "значение2"}.
- Множества (**set**): Набор уникальных значений, например, {1, 2, 3}.

Важно понимать, сколько памяти занимают различные типы данных, как они используются, и какие операции можно с ними выполнять.

1. Сколько памяти занимают переменные?
Размер памяти, который занимает переменная, зависит от ее типа и содержимого. Вот примерные размеры памяти для некоторых типов данных в Python:

Строки (str): Размер строки зависит от количества символов. Каждый символ в строке занимает 1 или 2 байта памяти (в зависимости от кодировки).
Булевы значения (bool): Занимают 1 байт памяти.
Списки (list): Размер списка зависит от количества элементов. Список занимает больше памяти, чем кортежи, так как списки являются изменяемыми структурами данных.
Кортежи (tuple): Кортежи занимают меньше памяти, чем списки, так как они неизменяемы.
Словари (dict): Словари занимают больше памяти, так как они хранят пары "ключ-значение".
Множества (set): Занимают меньше памяти по сравнению со словарями, так как в них хранятся только уникальные значения без пар "ключ-значение".

## 2. Использование типов данных
### Строки (str)
Строки используются для хранения текстовой информации, например, имен, адресов, сообщений и других данных, которые можно представить в виде текста.
Строки неизменяемы, что означает, что после создания строка не может быть изменена.
* Операции:
    - Конкатенация (объединение): "Hello" + " World!" -> "Hello, World!"
    - Повторение: "Hello" * 3 -> "HelloHelloHello"
    - Индексация и срезы: "Hello"[1] -> "e", "Hello"[1:4] -> "ell"
    - Проверка вхождения: "Hello" in "Hello, World!" -> True
    - Преобразование регистра: "Hello".upper() -> "HELLO", "Hello".lower() -> "hello"

### Булевы значения (bool)
Булевы значения используются для представления логических состояний (истина/ложь), они часто применяются в условных операторах и циклах.
* Операции:
    - Логическое "и" (and): True and False -> False
    - Логическое "или" (or): True or False -> True
    - Логическое "не" (not): not True -> False
    - Сравнение: 5 > 3 -> True, 2 == 2 -> True

    
### Списки (list)
Списки используются для хранения последовательности значений. Списки могут содержать элементы разных типов и являются изменяемыми (могут изменяться после создания).
Используются для хранения наборов данных, которые могут изменяться, например, списки студентов, продуктов, чисел и т.д.

* Операции:
    - Добавление элемента: my_list.append(4) -> [1, 2, 3, 4]
    - Удаление элемента: my_list.remove(2) -> [1, 3, 4]
    - Доступ по индексу: my_list[0] -> 1
    - Срезы: my_list[1:3] -> [2, 3]
    - Объединение списков: [1, 2] + [3, 4] -> [1, 2, 3, 4]

### Кортежи (tuple)
Кортежи похожи на списки, но они неизменяемы. Кортежи используются, когда требуется защитить данные от изменений.
Например, кортежи могут использоваться для хранения координат или других данных, которые не должны изменяться.

* Операции:
    - Доступ по индексу: my_tuple[0] -> 1
    - Срезы: my_tuple[1:3] -> (2, 3)
    - Объединение кортежей: (1, 2) + (3, 4) -> (1, 2, 3, 4)
    - Проверка вхождения: 2 in my_tuple -> True

* Основные особенности:
  - **Неизменяемость (иммутабельность)**: Кортежи неизменяемы, что означает, что после создания их содержимое нельзя изменить — ни добавить, ни удалить элементы, ни изменить их значения. Это делает кортежи идеальными для хранения данных, которые должны оставаться неизменными на протяжении выполнения программы. Пример: 
```python
my_tuple = (1, 2, 3)
# Попытка изменить значение вызовет ошибку
# my_tuple[0] = 10  # TypeError: 'tuple' object does not support item assignment
```
 - **Меньший объем памяти**: Поскольку кортежи неизменяемы, они занимают меньше памяти по сравнению со списками (list), которые являются изменяемыми. Это может быть важным фактором, когда необходимо оптимизировать использование памяти, например, при работе с большими наборами данных.
 - **Быстродействие**: Операции с кортежами выполняются быстрее, чем аналогичные операции со списками, из-за их неизменяемости. Это делает кортежи предпочтительными, когда требуется максимально быстрое выполнение программы.
 - **Использование в качестве ключей словаря и элементов множества**: В отличие от списков, кортежи могут использоваться в качестве ключей словаря и элементов множества, так как они неизменяемы и хэшируемы. Это делает их полезными для хранения сложных структур данных в словарях и множествах. Пример:
```python
my_dict = {('a', 1): "value1", ('b', 2): "value2"}
```
 - **Безопасность данных**: Неизменяемость кортежей обеспечивает дополнительную безопасность данных, так как исключает случайное или намеренное изменение значений. Это полезно в ситуациях, когда требуется передать набор данных в функцию или другую часть программы и быть уверенным, что данные останутся неизменными.
   
 - **Когда использовать кортежи**:
    + Когда данные не должны изменяться после создания.
    + Когда нужно использовать набор данных в качестве ключа словаря или элемента множества.
    + Когда важна оптимизация памяти и быстродействие.

### Словари (dict)

Словари используются для хранения данных в формате "ключ-значение". Они обеспечивают быстрый доступ к значениям по ключу.
Например, можно использовать словари для хранения контактной информации, где имя является ключом, а номер телефона — значением.
* Операции:
    - Доступ по ключу: my_dict["key1"] -> "value1"
    - Добавление пары "ключ-значение": my_dict["key3"] = "value3" -> {"key1": "value1", "key2": "value2", "key3": "value3"}
    - Удаление пары: del my_dict["key2"] -> {"key1": "value1", "key3": "value3"}
    - Получение всех ключей: my_dict.keys() -> dict_keys(['key1', 'key3'])
    - Проверка наличия ключа: "key1" in my_dict -> True

* **Основные особенности**:
 - **Хранение данных в формате "ключ-значение"**: Словари хранят данные в виде пар "ключ-значение". Это делает словари чрезвычайно удобными для быстрого поиска значений по уникальному ключу.
Пример:
```python
my_dict = {"name": "Alice", "age": 25, "city": "New York"}
print(my_dict["name"])  # "Alice"
```
 - **Изменяемость (мутабельность)**: Словари являются изменяемыми, что означает, что после создания словаря можно добавлять, удалять и изменять элементы. Это делает словари гибкими и удобными для работы с динамическими данными.
Пример:
```python
my_dict = {"name": "Alice"}
my_dict["age"] = 25  # Добавление нового элемента
my_dict["name"] = "Bob"  # Изменение существующего элемента
```

 - **Уникальные ключи**: В словаре каждый ключ должен быть уникальным. Если в словарь добавляется пара с ключом, который уже существует, старое значение будет заменено новым.
Пример:
```python
my_dict = {"name": "Alice", "name": "Bob"}
print(my_dict["name"])  # "Bob"
```

* **Быстрый доступ к данным**: Словари обеспечивают быстрый доступ к значениям по ключу благодаря использованию хеш-таблиц. Это делает операции поиска, добавления и удаления элементов очень эффективными по времени (в среднем O(1)).
 - **Поддержка вложенных структур данных**: Словари могут содержать другие словари или сложные структуры данных в качестве значений, что позволяет создавать сложные иерархические структуры для хранения данных.
Пример:
```python
my_dict = {
    "name": "Alice",
    "details": {
        "age": 25,
        "city": "New York"
    }
}
print(my_dict["details"]["age"])  # 25
```

* **Гибкость и функциональность**: Словари предоставляют множество методов для работы с данными, такие как get(), pop(), items(), keys(), values() и другие, что делает их мощным инструментом для управления данными.
 - **Когда использовать словари**:
   + Когда необходимо хранить данные с уникальными ключами и быстро получать доступ к значениям по ключу.
   + Когда данные могут изменяться и требуется гибкость для добавления и удаления элементов.
   + Когда нужно организовать данные в виде пар "ключ-значение", особенно если ключи уникальны и хэшируемы.
   + Когда требуется использовать сложные вложенные структуры данных для организации информации.

### Множества (set)

Множества используются для хранения уникальных элементов, например, уникальных значений в списке или множества категорий.
Например, множество может использоваться для удаления дубликатов из списка.
* Операции:

    - Добавление элемента: my_set.add(4) -> {1, 2, 3, 4}
    - Удаление элемента: my_set.remove(2) -> {1, 3, 4}
    - Объединение множеств: my_set | another_set -> объединение всех элементов
    - Пересечение множеств: my_set & another_set -> элементы, присутствующие в обоих множествах
    - Проверка вхождения: 2 in my_set -> True

### Методы для работы со строками (str)

**Строки в Python** - это неизменяемые последовательности символов. Несмотря на это, существует множество методов, которые позволяют работать с ними эффективно.

**```str.upper()```**: Преобразует все символы строки в верхний регистр.

````python
text = "hello"
print(text.upper())  # "HELLO"
print(text)
````

**```str.lower()```**: Преобразует все символы строки в нижний регистр.

````python
text = "HELLO"
print(text.lower())  # "hello"
````

**```str.capitalize()```**: Преобразует первую букву строки в верхний регистр.

````python
text = "hello world"
print(text.capitalize())  # "Hello world"
````

**```str.find(substring)```**: Возвращает индекс первого вхождения подстроки. Если подстрока не найдена, возвращает -1.

````python
text = "hello world"
print(text.find("world"))  # 6
````

**```str.replace(old, new)```**: Заменяет все вхождения подстроки old на new.

````python
text = "hello world"
print(text.replace("world", "Python"))  # "hello Python"
````

**```str.split(delimiter)```**: Разделяет строку по указанному разделителю и возвращает список.

````python
text = "apple,banana,grape"
print(text.split(","))  # ['apple', 'banana', 'grape']
````

**```str.join(iterable)```**: Соединяет элементы в переданном итерируемом объекте (например, список) в одну строку, используя строку в качестве разделителя.

````python
fruits = ["apple", "banana", "grape"]
print(", ".join(fruits))  # "apple, banana, grape"
````

### Методы для работы со списками (list)
Списки в Python - это изменяемые последовательности данных, которые могут содержать элементы различных типов.

**```list.append(item)```**: Добавляет элемент в конец списка.

````python
numbers = [1, 2, 3]
numbers.append(4)
print(numbers)  # [1, 2, 3, 4]
````

**```list.extend(iterable)```**: Расширяет список, добавляя элементы из переданного итерируемого объекта (например, из другого списка).

````python
numbers = [1, 2, 3]
numbers.extend([4, 5])
print(numbers)  # [1, 2, 3, 4, 5]
````

**```list.insert(index, item)```**: Вставляет элемент в список по указанному индексу.

````python
numbers = [1, 2, 3]
numbers.insert(1, 10)
print(numbers)  # [1, 10, 2, 3]
````

**```list.remove(item)```**: Удаляет первое вхождение указанного элемента из списка.

````python
numbers = [1, 2, 3, 2]
numbers.remove(2)
print(numbers)  # [1, 3, 2]
````

**```list.pop(index)```**: Удаляет и возвращает элемент по указанному индексу (по умолчанию удаляет и возвращает последний элемент).

````python
numbers = [1, 2, 3]
print(numbers.pop())  # 3
print(numbers)  # [1, 2]
````

**```list.index(item)```**: Возвращает индекс первого вхождения элемента в списке.

````python
numbers = [1, 2, 3, 2]
print(numbers.index(2))  # 1
````

**```list.count(item)```**: Возвращает количество вхождений элемента в список.

````python
numbers = [1, 2, 3, 2]
print(numbers.count(2))  # 2
````

**```list.sort()```**: Сортирует список на месте.

````python
numbers = [3, 1, 2]
numbers.sort()
print(numbers)  # [1, 2, 3]
````

**```list.reverse()```**: Переворачивает список на месте.

````python
numbers = [1, 2, 3]
numbers.reverse()
print(numbers)  # [3, 2, 1]
````

### Методы для работы с кортежами (tuple)
Кортежи в Python - это неизменяемые последовательности данных. Хотя кортежи поддерживают те же операции, что и списки, они не имеют методов изменения данных (например, append() или remove()), так как кортежи неизменяемы.

**```tuple.count(item)```**: Возвращает количество вхождений элемента в кортеж.

````python
my_tuple = (1, 2, 3, 2)
print(my_tuple.count(2))  # 2
````

**```tuple.index(item)```**: Возвращает индекс первого вхождения элемента в кортеже.

````python
my_tuple = (1, 2, 3)
print(my_tuple.index(2))  # 1
````

### Методы для работы со словарями (dict)
Словари в Python - это изменяемые структуры данных, которые хранят пары "ключ-значение".

**```dict.keys()```**: Возвращает все ключи словаря.

````python
my_dict = {"key1": "value1", "key2": "value2"}
print(my_dict.keys())  # dict_keys(['key1', 'key2'])
````

**```dict.values()```**: Возвращает все значения словаря.

````python
my_dict = {"key1": "value1", "key2": "value2"}
print(my_dict.values())  # dict_values(['value1', 'value2'])
````

**```dict.items()```**: Возвращает пары "ключ-значение" в виде кортежей.

````python
my_dict = {"key1": "value1", "key2": "value2"}
print(my_dict.items())  # dict_items([('key1', 'value1'), ('key2', 'value2')])
````

**```dict.get(key, default)```**: Возвращает значение по ключу. Если ключ не найден, возвращает значение по умолчанию.

````python
my_dict = {"key1": "value1"}
print(my_dict.get("key1"))  # "value1"
print(my_dict.get("key2", "default_value"))  # "default_value"
````

**```dict.pop(key)```**: Удаляет и возвращает значение по ключу. Если ключ не найден, выбрасывает исключение KeyError.

````python
my_dict = {"key1": "value1"}
print(my_dict.pop("key1"))  # "value1"
print(my_dict)  # {}
````

**```dict.update(other_dict)```**: Обновляет словарь, добавляя пары "ключ-значение" из другого словаря.

````python
my_dict = {"key1": "value1"}
my_dict.update({"key2": "value2"})
print(my_dict)  # {"key1": "value1", "key2": "value2"}
````

### Методы для работы с множествами (set)
Множества в Python — это изменяемые структуры данных, которые хранят уникальные элементы.

**```set.add(item)```**: Добавляет элемент во множество.

````python
my_set = {1, 2, 3}
my_set.add(4)
print(my_set)  # {1, 2, 3, 4}
````

**```set.remove(item)```**: Удаляет элемент из множества. Если элемент не найден, выбрасывает исключение KeyError.

````python
my_set = {1, 2, 3}
my_set.remove(2)
print(my_set)  # {1, 3}
````

**```set.discard(item)```**: Удаляет элемент из множества, если он присутствует. Если элемент не найден, ничего не происходит.

````python
my_set = {1, 2, 3}
my_set.discard(2)
my_set.discard(4)  # Никакого исключения не будет
print(my_set)  # {1, 3}
````

**```set.pop()```**: Удаляет и возвращает случайный элемент из множества. Если множество пусто, выбрасывается исключение KeyError.

````python
my_set = {1, 2, 3}
element = my_set.pop()
print(element)  # Может вернуть любой элемент, например, 1
print(my_set)  # Остальные элементы, например, {2, 3}
````

**```set.clear()```**: Очищает множество, удаляя все элементы.

````python
my_set = {1, 2, 3}
my_set.clear()
print(my_set)  # set()
````

**```set.union(other_set)```**: Возвращает новое множество, содержащее объединение двух множеств.

````python
set1 = {1, 2, 3}
set2 = {3, 4, 5}
union_set = set1.union(set2)
print(union_set)  # {1, 2, 3, 4, 5}
````

**```set.intersection(other_set)```**: Возвращает новое множество, содержащее пересечение двух множеств.

````python
set1 = {1, 2, 3}
set2 = {2, 3, 4}
intersection_set = set1.intersection(set2)
print(intersection_set)  # {2, 3}
````

**```set.difference(other_set)```**: Возвращает новое множество, содержащее разницу между двумя множествами.

````python
set1 = {1, 2, 3}
set2 = {2, 3, 4}
difference_set = set1.difference(set2)
print(difference_set)  # {1}
````

**```set.symmetric_difference(other_set)```**: Возвращает новое множество, содержащее элементы, которые есть в одном из множеств, но не в обоих.

````python
set1 = {1, 2, 3}
set2 = {2, 3, 4}
symmetric_difference_set = set1.symmetric_difference(set2)
print(symmetric_difference_set)  # {1, 4}
````

## Примеры использования циклов для работы со списками и словарями

### 1. Цикл for для работы со списком
Один из самых распространенных способов работы со списками - это использование цикла for, который позволяет последовательно проходить по каждому элементу списка.

Пример:

````python
numbers = [1, 2, 3, 4, 5]

# Увеличим каждый элемент списка на 1
for num in numbers:
    print(num + 1)
````

### Объяснение: 
Здесь цикл for проходит по каждому элементу списка numbers. Внутри цикла мы увеличиваем значение элемента на 1 и выводим результат. 
Цикл for удобен тем, что автоматически перебирает элементы списка, и вам не нужно заботиться об индексах.

### 2. Цикл for с использованием индексов
Иногда нужно не просто получить значение элемента, но и его индекс. Для этого удобно использовать функцию enumerate().

Пример:

````python
fruits = ["apple", "banana", "cherry"]

# Добавляем к каждому фрукту его индекс в списке
for index, fruit in enumerate(fruits):
    print(f"{index + 1}: {fruit}")
````

**Объяснение**: Функция ```enumerate()``` возвращает пары (индекс, значение), что позволяет вам получить и индекс, и сам элемент. Это полезно, когда вы хотите работать с элементами списка с учетом их позиции.

### 3. Цикл for с использованием списка кортежей
Иногда список может содержать сложные элементы, такие как кортежи. Вы можете распаковывать эти элементы прямо в цикле.

Пример:

````python
students = [("Alice", 20), ("Bob", 22), ("Charlie", 19)]

# Выводим имена студентов и их возраст
for name, age in students:
    print(f"Name: {name}, Age: {age}")
````

**Объяснение**: В данном примере каждый элемент списка ```students``` - это кортеж из двух значений (имя, возраст). В цикле ```for``` мы распаковываем их прямо в переменные ```name``` и ```age``` и затем используем эти переменные в теле цикла.

### 4. Цикл for для работы со словарем
Цикл ```for``` также может быть использован для работы со словарями. Можно итерироваться по ключам, значениям или сразу по парам ключ-значение.

Пример:

````python
student_ages = {"Alice": 20, "Bob": 22, "Charlie": 19}

# Проходим по парам ключ-значение
for name, age in student_ages.items():
    print(f"{name} is {age} years old")
````

**Объяснение**: Метод ```.items()``` возвращает пары (ключ, значение) для словаря, что позволяет вам работать сразу с обеими частями данных. Этот подход удобен, когда нужно выполнить какие-то операции как с ключами, так и с значениями.

### 5. Цикл for для изменения элементов списка
Если нужно изменить элементы списка на месте, можно использовать цикл ```for``` с индексацией.

Пример:

````python
numbers = [1, 2, 3, 4, 5]

# Умножаем каждый элемент списка на 2
for i in range(len(numbers)):
    numbers[i] *= 2

print(numbers)
````

**Объяснение**: Здесь мы используем функцию ```range()```, которая генерирует последовательность чисел от 0 до длины списка (не включая последний элемент). Это позволяет нам итерироваться по индексам и изменять элементы списка на месте.

## 6. Цикл while для работы со списками
Цикл ```while``` также может быть использован для работы со списками, особенно когда нужно контролировать процесс итерации более гибко.

Пример:

````python
numbers = [5, 10, 15, 20]
index = 0

# Умножаем каждый элемент на 3 до тех пор, пока не дойдем до конца списка
while index < len(numbers):
    numbers[index] *= 3
    index += 1

print(numbers)
````

**Объяснение**: Цикл ```while``` работает до тех пор, пока выполняется условие ```index < len(numbers)```. Внутри цикла мы изменяем элементы списка и увеличиваем индекс для перехода к следующему элементу.

Циклы в Python предоставляют мощный и гибкий инструмент для работы со списками и словарями. Использование циклов for и while позволяет выполнять самые разные задачи: от простого перебора элементов до сложных операций с индексами и ключами. Понимание особенностей и вариантов использования циклов помогает писать более эффективный и понятный код.

## Встроенные функции для работы с циклами (range, enumerate, zip).

### 1. Функция range()
Функция **range()** используется для генерации последовательности чисел, которые могут быть использованы в циклах **for**. Она полезна, когда вам нужно повторить какую-то операцию определенное количество раз или итерироваться по индексам списка.

- Синтаксис:
    + range(start, stop, step)
    + start (необязательный): начальное значение (включительно). По умолчанию 0.
    + stop: конечное значение (исключительно).
    + step (необязательный): шаг между значениями. По умолчанию 1.

Примеры использования:

````python
# Проход по числам от 0 до 4
for i in range(5):
    print(i)
````

````python
# Проход по четным числам от 2 до 8
for i in range(2, 10, 1):
    print(i)
````

## 2. Функция enumerate()

Функция **enumerate()** используется для итерации по элементам последовательности с одновременным получением их индексов. Это особенно полезно, когда вам нужно итерироваться по списку или другому итерируемому объекту и знать позиции элементов.

 - Синтаксис:
    + enumerate(iterable, start=0)
    + iterable: объект, по которому будет происходить итерация.
    + start (необязательный): начальное значение для индексации. По умолчанию 0.

Пример использования:

````python
fruits = ["apple", "banana", "cherry"]

for index, fruit in enumerate(fruits):
    print(f"{index}: {fruit}")
````

## 3. Функция zip()

Функция zip() используется для объединения нескольких итерируемых объектов (например, списков) в один итерируемый объект, состоящий из кортежей. Каждый кортеж будет содержать элементы с соответствующими индексами из всех итерируемых объектов.

- Синтаксис:
    + zip(*iterables). *iterables: два или более итерируемых объекта.

Пример использования:

````python
names = ["Alice", "Bob", "Charlie"]
ages = [25, 30, 35]

for name, age in zip(names, ages):
    print(f"{name} is {age} years old")
````

## Срезы

**Срез** - это часть последовательности. Работает одинаково для строк, списков
и кортежей.

- `s[a:b]` - от элемента `a` включительно до `b` **не включая**;
- `s[a:]` - от `a` до конца;
- `s[:b]` - от начала до `b`;
- `s[a:b:step]` - с шагом;
- отрицательные индексы считаются с конца: `s[-1]` - последний элемент.

Главное правило: **правая граница не входит в срез**. Поэтому `s[0:3]` даёт
ровно три элемента, а `s[:3] + s[3:]` восстанавливает исходную
последовательность без потерь и повторов.

````python
text = "Учёные открыли новый вид бактерий"

print(text[:7])        # 'Учёные ' — первые семь символов
print(text[-9:])       # последние девять символов
print(text[::2])       # каждый второй символ
print(text[::-1])      # строка задом наперёд

numbers = [10, 20, 30, 40, 50]
print(numbers[1:4])    # [20, 30, 40]
print(numbers[:2] + numbers[2:])   # исходный список: границы стыкуются без потерь
print(numbers[-2:])    # два последних
````

> Срез **никогда не вызывает ошибку выхода за границы**. `numbers[10:20]`
> на списке из пяти элементов вернёт пустой список, а не исключение.
> Это удобно и опасно одновременно: опечатка в границах не падает, а молча
> возвращает пустоту.

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[10:20])   # [] — пусто, но без ошибки
print(numbers[3:1])     # [] — начало правее конца

# а вот обращение по индексу за границей — уже ошибка:
try:
    print(numbers[10])
except IndexError as error:
    print("IndexError:", error)
````

## Сортировка по правилу: `key` и `lambda`

По умолчанию `sorted` сравнивает элементы целиком. Часто нужно сортировать
не по самому элементу, а по какой-то его части - для этого есть параметр `key`.

`key` - это **функция**, которую применят к каждому элементу перед сравнением.
Писать её отдельно ради одной строки неудобно, поэтому пользуются `lambda` -
безымянной функцией из одного выражения:

```python
lambda x: x[1]      # «взять у элемента второй компонент»
```

`reverse=True` переворачивает порядок. Те же `key` и `reverse` понимают
`min()` и `max()`.

````python
counts = [("новость", 5), ("учёные", 12), ("вирус", 7), ("россия", 12)]

print("по алфавиту      :", sorted(counts))
print("по числу         :", sorted(counts, key=lambda pair: pair[1]))
print("по числу, убывая :", sorted(counts, key=lambda pair: pair[1], reverse=True))

print()
words = ["кот", "бактерия", "вирус", "ёж"]
print("самое длинное слово:", max(words, key=len))
print("самое короткое     :", min(words, key=len))
print("по длине           :", sorted(words, key=len))

# частый приём: сортировка по двум ключам сразу —
# сначала по числу (убывая), при равенстве по алфавиту
print()
print(sorted(counts, key=lambda pair: (-pair[1], pair[0])))
````

## Словарь: `get` и `setdefault`

Обращение к несуществующему ключу через квадратные скобки - ошибка `KeyError`.
Два метода позволяют её избежать:

- **`d.get(key, default)`** - вернуть значение или `default`, если ключа нет.

- **`d.setdefault(key, default)`** - вернуть значение, а если ключа нет -

Первый нужен для чтения, второй - для накопления. На них строятся два

### Приём 1: частотный словарь

````python
words = ["кот", "пёс", "кот", "ёж", "кот", "пёс"]

frequency = {}
for word in words:
    frequency[word] = frequency.get(word, 0) + 1

print(frequency)
print("самое частое:", max(frequency, key=frequency.get))
print("топ по убыванию:", sorted(frequency.items(), key=lambda pair: -pair[1]))
````

## Сопоставление с образцом: `match` / `case`

Когда вариантов много, цепочка `if / elif / elif / else` становится длинной
и однообразной. С Python 3.10 для таких случаев есть `match`:

```python
match значение:
    case образец_1:
        ...
    case образец_2:
        ...
    case _:
        ...
```

Значение сверху по очереди сравнивается с образцами; **выполняется первая
подходящая ветка**, остальные пропускаются. `case _` - «всё остальное»,
аналог `else`; его пишут последним.

Это не «более модный if». Разница в том, что `case` сравнивает не только
значения, но и **форму** данных - длину кортежа, наличие ключей, - и сразу
раскладывает части по переменным.

````python
def rating_label(rating):
    match rating:
        case 5:
            return "отлично"
        case 4:
            return "хорошо"
        case 3:
            return "средне"
        case 1 | 2:                 # несколько значений через |
            return "плохо"
        case _:                     # всё остальное
            return "неизвестная оценка"


for value in (5, 4, 3, 2, 1, 99):
    print(value, "->", rating_label(value))
````

### То же самое через `if` - для сравнения

`match` не даёт ничего такого, чего нельзя написать через `if`. Он даёт
читаемость: в варианте ниже слово `rating` повторяется пять раз, и в одной
из строк его легко перепутать с чем-нибудь похожим.

````python
def rating_label_if(rating):
    if rating == 5:
        return "отлично"
    elif rating == 4:
        return "хорошо"
    elif rating == 3:
        return "средне"
    elif rating == 1 or rating == 2:
        return "плохо"
    else:
        return "неизвестная оценка"


print(all(rating_label(v) == rating_label_if(v) for v in (1, 2, 3, 4, 5, 99)))
````

### Условие внутри ветки: `case ... if ...`

К образцу можно добавить дополнительное условие - его называют «сторож».
Ветка сработает, только если и образец подошёл, и условие истинно.

````python
def length_label(words):
    match words:
        case n if n < 20:
            return "короткий"
        case n if n < 50:
            return "обычный"
        case _:
            return "длинный"


for value in (10, 34, 61):
    print(value, "слов ->", length_label(value))
````

### Сопоставление формы: кортежи и списки

Вот здесь `match` делает то, чего `if` не умеет: проверяет **структуру**
значения и одновременно раскладывает его части по именам.

````python
def describe(record):
    match record:
        case (name, count) if count == 0:
            return f"{name}: записей нет"
        case (name, count):
            return f"{name}: {count} записей"
        case [single]:
            return f"единственный элемент: {single}"
        case _:
            return "неизвестный формат"


print(describe(("книги", 312)))
print(describe(("одежда", 0)))
print(describe(["всё"]))
print(describe(42))
````

## Списковые включения

Цикл, который строит список, записывается одной строкой:

```python
[выражение for элемент in последовательность if условие]
```

Читается слева направо: «выражение - для каждого элемента - если условие».
Часть с `if` необязательна.

Это не «более красивый цикл», а другая запись того же самого. Пользоваться
ею стоит, когда цикл **строит список и больше ничего не делает**; если внутри
несколько действий, обычный `for` читается лучше.

````python
words = ["Кот", "ПЁС", "ёж", "Бактерия", "вирус"]

# обычный цикл
lower = []
for word in words:
    lower.append(word.lower())
print(lower)

# то же самое включением
print([word.lower() for word in words])

# с условием: только длинные слова
print([word for word in words if len(word) > 3])

# выражение и условие вместе
print([len(word) for word in words if word.isalpha()])

# включения бывают и для словарей, и для множеств
print({word: len(word) for word in words})
print({len(word) for word in words})
````

## Пример 1. Счётчик в цикле с условием

Самая частая пара «цикл + условие»: пройти по данным и посчитать, сколько
элементов удовлетворяет условию.

Три способа записать одно и то же - от развёрнутого к короткому.

````python
# способ 1: явный счётчик
long_texts = 0
for text in TEXTS:
    if len(text.split()) > 8:
        long_texts += 1
print("длинных новостей:", long_texts)

# способ 2: включение и len
print("длинных новостей:", len([t for t in TEXTS if len(t.split()) > 8]))

# способ 3: sum по условию — True считается за 1
print("длинных новостей:", sum(1 for t in TEXTS if len(t.split()) > 8))

# и доля вместо количества
share = sum(1 for t in TEXTS if len(t.split()) > 8) / len(TEXTS)
print(f"доля длинных: {share:.0%}")
````

## Пример 2. Накопление суммы и среднего

Второй по частоте приём: пройти по данным, накопить сумму, поделить на количество.

````python
lengths = []
for text in TEXTS:
    lengths.append(len(text.split()))

print("длины:", lengths)
print("всего слов:", sum(lengths))
print(f"средняя длина: {sum(lengths) / len(lengths):.2f}")

# то же самое одной строкой
lengths = [len(text.split()) for text in TEXTS]
print(f"средняя длина: {sum(lengths) / len(lengths):.2f}")
````

## Пример 3. Частотный словарь по тексту

Классическая задача обработки текста: посчитать, сколько раз встречается
каждое слово. Здесь встречаются сразу четыре приёма: цикл по списку,
приведение к нижнему регистру, `get` с умолчанием и сортировка по значению.

````python
frequency = {}
for text in TEXTS:
    for word in text.lower().split():
        word = word.strip(".,:!?")          # убираем знаки по краям
        if len(word) < 3:                   # короткие слова пропускаем
            continue
        frequency[word] = frequency.get(word, 0) + 1

print("уникальных слов:", len(frequency))

top = sorted(frequency.items(), key=lambda pair: (-pair[1], pair[0]))
print("топ-5:")
for word, count in top[:5]:
    print(f"   {word.ljust(14)} {count}")
````

Посмотрите на топ внимательно: среди слов оказался токен **`3:1`**.
Метод `.strip(".,:!?")` убирает знаки только **по краям** строки, а двоеточие
внутри `3:1` остаётся на месте - и счёт матча становится «словом».

Это не недоработка примера, а настоящий вопрос: что считать словом. В практике
вам предстоит ответить на него самому, и от ответа изменится число уникальных
слов. Никакая функция не решит его за вас - это решение, а не вычисление.

> Здесь пригодились два оператора управления циклом. **`continue`**
> пропускает оставшуюся часть тела и переходит к следующему элементу.
> **`break`** прерывает цикл целиком - он понадобится в примере 7.

## Пример 4. Группировка по ключу

Задача, которая в анализе данных встречается чаще любой другой: разложить
записи по группам и посчитать что-нибудь в каждой.

Сначала собираем словарь списков, потом считаем по каждому списку.

````python
# шаг 1: раскладываем оценки по категориям
by_category = {}
for category, rating, words in ROWS:
    by_category.setdefault(category, []).append(rating)

print(by_category)

# шаг 2: считаем среднее в каждой группе
print()
print("категория".ljust(16) + "n".rjust(3) + "среднее".rjust(10))
for category, ratings in by_category.items():
    print(category.ljust(16) + str(len(ratings)).rjust(3)
          + f"{sum(ratings) / len(ratings):.2f}".rjust(10))

# шаг 3: сортируем группы по среднему
print()
summary = [(name, len(v), sum(v) / len(v)) for name, v in by_category.items()]
for name, count, value in sorted(summary, key=lambda row: -row[2]):
    print(f"{name.ljust(16)} n={count}  {value:.2f}")
````

## Пример 5. Два списка параллельно

Когда данные лежат в нескольких списках с совпадающими позициями, их
проходят вместе через `zip`. Это надёжнее, чем цикл по индексам: `zip`
не даст ошибиться на единицу и сам остановится на более коротком списке.

````python
# по индексам — работает, но легко ошибиться
for index in range(len(TEXTS)):
    print(TOPICS[index], "|", TEXTS[index][:40])

print()
# через zip — короче и безопаснее
for topic, text in zip(TOPICS, TEXTS):
    print(topic, "|", text[:40])

print()
# zip останавливается на самом коротком
print(list(zip([1, 2, 3], ["а", "б"])))

print()
# группировка через zip: средняя длина новости по теме
lengths_by_topic = {}
for topic, text in zip(TOPICS, TEXTS):
    lengths_by_topic.setdefault(topic, []).append(len(text.split()))

for topic, values in lengths_by_topic.items():
    print(f"{topic.ljust(8)} {sum(values) / len(values):.1f} слова в среднем")
````

## Пример 6. Поиск максимума с условием

`max()` находит наибольший элемент, но не говорит, **где** он. Если нужен
и сам элемент, и его позиция, цикл с ручным сравнением остаётся самым
понятным способом.

````python
# через max с ключом - коротко
longest = max(TEXTS, key=lambda text: len(text.split()))
print("самая длинная новость:", longest[:45])

# циклом - когда нужен ещё и индекс
best_index = 0
best_length = -1
for index, text in enumerate(TEXTS):
    current = len(text.split())
    if current > best_length:
        best_length, best_index = current, index

print(f"её номер: {best_index}, слов: {best_length}")

# максимум среди тех, что удовлетворяют условию
science = [t for t, topic in zip(TEXTS, TOPICS) if topic == "наука"]
print("самая длинная научная:", max(science, key=len)[:45])
````

## Пример 7. Цикл `while` и `break`

`for` проходит по готовой последовательности. `while` работает, пока верно
условие, - им пользуются, когда заранее неизвестно, сколько шагов понадобится.

Типичный случай: читать элементы, пока не встретится стоп-значение.

````python
stream = ["первая новость", "вторая новость", "СТОП", "третья новость"]

collected = []
index = 0
while index < len(stream):
    item = stream[index]
    if item == "СТОП":
        print("встретили стоп-слово, дальше не читаем")
        break
    collected.append(item)
    index += 1

print("собрано:", collected)

# тот же приём через for - обычно короче
collected = []
for item in stream:
    if item == "СТОП":
        break
    collected.append(item)
print("собрано:", collected)
````

> **Осторожно с `while`.** Если условие никогда не станет ложным, цикл
> не закончится. В примере выше `index += 1` обязателен: забудете -
> программа зависнет. Именно поэтому `for` предпочтительнее везде,
> где число шагов известно заранее.

## Пример 8. Множества: общее и различное

Множество хранит только уникальные значения и умеет сравниваться с другим
множеством. На этом строится сравнение двух текстов: какие слова есть
в обоих, какие только в одном.

````python
first = set(TEXTS[0].lower().split())
second = set(TEXTS[2].lower().split())

print("слов в первой :", len(first))
print("слов в третьей:", len(second))
print()
print("общие     (&):", sorted(first & second))
print("только в 1 (-):", sorted(first - second))
print("все вместе (|):", len(first | second))
print()
# доля общих слов — простейшая мера похожести двух текстов
similarity = len(first & second) / len(first | second)
print(f"похожесть: {similarity:.2f}")
````

## Пример 9. Вложенные циклы

Цикл внутри цикла нужен, когда каждый элемент надо сравнить с каждым.
Помните про цену: на 100 элементах это 10 000 сравнений, на 1000 - миллион.

````python
# похожесть каждой новости с каждой
sets = [set(text.lower().split()) for text in TEXTS]

print("    " + "".join(f"{i:>7}" for i in range(len(sets))))
for i, left in enumerate(sets):
    row = f"{i:>3} "
    for right in sets:
        value = len(left & right) / len(left | right)
        row += f"{value:>7.2f}"
    print(row)

print()
print("по диагонали единицы: текст всегда полностью похож на себя")
print(f"сравнений сделано: {len(sets)} x {len(sets)} = {len(sets) ** 2}")
````

## Пример 10. Отчёт: всё вместе

Финальный пример собирает приёмы из всех предыдущих: цикл с условием,
накопление, группировку, сортировку и выравнивание при выводе.
Ровно так будет устроена последняя задача практики.

````python
# считаем показатели
total_words = sum(len(text.split()) for text in TEXTS)
unique_words = len({w.lower().strip(".,:!?") for text in TEXTS for w in text.split()})
by_topic = {}
for topic, text in zip(TOPICS, TEXTS):
    by_topic.setdefault(topic, []).append(len(text.split()))

# печатаем отчёт
lines = [
    ("новостей", len(TEXTS)),
    ("слов всего", total_words),
    ("уникальных слов", unique_words),
    ("средняя длина", f"{total_words / len(TEXTS):.1f} слова"),
    ("тем", len(by_topic)),
]

print("ОТЧЁТ ПО ПОДБОРКЕ")
print("-" * 34)
for name, value in lines:
    print(name.ljust(20) + ":" + str(value).rjust(13))

print()
print("по темам:")
for topic, values in sorted(by_topic.items(), key=lambda pair: -len(pair[1])):
    print(f"   {topic.ljust(8)} новостей {len(values)}, "
          f"средняя длина {sum(values) / len(values):.1f}")
````
