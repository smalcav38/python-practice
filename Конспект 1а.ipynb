# 1. Тетрадь, `print` и первая ошибка

Программа на Python - это последовательность команд, которые выполняются сверху вниз. В блокноте они разложены по **ячейкам**: ячейку можно запускать, она выполняется, а под ней появляется результат.

Две вещи, без которых дальше не обойтись:

- **`print(...)`** печатает то, что стоит в скобках. Значений может быть несколько через запятую, разделителем по умолчанию будет являться пробел.
- **`#`** начинает комментарий: всё до конца строки Python игнорирует.

```python
print("Привет")
print("Привет", "мир")

# это комментарий, он не выполняется
print("рубрика:", "наука", "| оценка:", 5)
```

### Ячейка печатает последнее выражение сама

В блокноте (**и только в нём**) есть особенность: если последняя строка ячейки - это выражение, его значение напечатается и без `print`. Промежуточные же строки печататься не будут.

Это удобно, однако код, который «работал в блокноте», в обычном файле ничего не выведет на экран. Поэтому в практике везде используют `print`.

```python
"первое значение"
"второе значение"
```

### Красный текст - это ошибка, и её надо понимать

Ошибка не ломает ничего и не портит код. Она сообщает, что именно не понравилось Python (чего именно он не понял).

Смотреть надо на **последнюю строку**, там описывается тип ошибки и её объяснение.
Всё, что выше, — путь, по которому Python шёл к этому месту.

Четыре ошибки, которые Вы встретите на первом занятии:

| Ошибка | Что значит |
| --- | --- |
| `NameError` | использовано имя, которого нет. Опечатка или забыли создать |
| `TypeError` | операция не подходит этим типам. Чаще всего - сложили строку с числом |
| `IndexError` | обращение к элементу, которого не существует |
| `SyntaxError` | Python не понимает что ему делать. Обычно не закрыта скобка или кавычка |

Раскомментируйте строки по одной и посмотрите на каждую.

```python
# print(privet)
# print("возраст: " + 5)
# print(TEXTS[10])
# print("забыли закрыть скобку"
```

> `SyntaxError` показывает **следующую** строку после ошибочной -
> потому что Python понимает, что что-то не так, только дойдя до неё.
> Если ошибка показана на строке, в которой всё в порядке, смотрите выше.

# 2. Переменная

Переменная - это имя для значения. Знак `=` читается не «равно», а **«положи в»**:
справа находится значение, слева находится имя, куда его присваивать.

```python
label = "наука"
score = 5

print(label, score)

label = "спорт"
print(label)
```

Главное, ради чего переменные необходимы: **в них кладут результат вычисления**, а не только то, что напечатали руками. Один раз посчитали - дальше используем.

```python
text = TEXTS[0]

chars = len(text)
words = len(text.split())

print(chars, words)
print("в записи", chars, "символов и", words, "слов")

chars = chars + 100
print(chars)
```

> Строка `chars = chars + 100` как математическое равенство - бессмыслица.
> Как команда — норма: «возьми то, что сейчас в `chars`, прибавь сто,
> положи обратно».

# 3. Четыре основных типа переменных

У каждого значения переменной есть тип. Объявлять его не нужно - Python автоматически определяет его. Это свойство называется **динамической типизацией**.

| Тип | Что это | Пример |
| --- | --- | --- |
| `str` | строка, текст | `"наука"` |
| `int` | целое число | `5` |
| `float` | число с дробной частью | `3.45` |
| `bool` | истина или ложь | `True`, `False` |

Посмотреть тип: `type(значение)`.

```python
print(type("наука"))
print(type(5))
print(type(3.45))
print(type(True))

print(type("5"))
```

Последняя строка - та самая ловушка. `5` и `"5"` выглядят похоже, однако первое — это число, а второе - это строка из одного символа. Числа складываются, строки склеиваются, и смешивать их нельзя.

Тип переменной может меняться по ходу программы - это удобно, однако иногда это может являться источником ошибок.

```python
value = 5
print(type(value))

value = "пять"
print(type(value))

# А вот и TypeError. Раскомментируйте:
# print(5 + "5")
```

# 4. Строка: длина, регистр, склейка, подстановка

**`len(x)`** - длина. Считаются все символы, включая пробелы и знаки препинания.

**`.upper()`, `.lower()`** - регистр. Запись через точку называется **методом**: «у этой строки выполни это действие». Скобки обязательны.

Важное свойство: метод **всегда возвращает новую строку**, а исходную не не изменяет.
Строки в Python неизменяемы, т.е. переписать букву внутри нельзя, можно только создать новую строку.

```python
text = TEXTS[1]

print(len(text))
print(text.upper())
print(text.lower())

print(text)

print(text.upper)
```

Последняя строка напечатала `<built-in method ...>` - это метод без скобок.
Ошибки нет. Забытые скобки - самая частая причина того, что «код работает, но печатает ерунду».

### Склейка и f-строка

Знак `+` между строками склеивает их. Пробел сам не появится, и число cклеить со строкой будет нельзя ( `TypeError` ).

Поэтому для вывода текста обычно используется **f-строка**: буква `f` перед начальной кавычкой, внутри фигурных скобок - имя переменной или любое выражение. Типы переменных при этом не важны.

```python
label = "наука"
score = 5

print(label + " — " + "запись")

print(f"{label}: оценка {score}")
print(f"удвоенная оценка: {score * 2}")
print(f"метка заглавными: {label.upper()}")

print(f"доля: {2 / 3:.0%}")
print(f"число: {5.6789:.2f}")
```

> После двоеточия внутри скобок задаётся формат: `:.2f` - два знака после
> запятой, `:.0%` - проценты, `:>8` - выровнять по правому краю в поле
> шириной восемь. Ту же задачу выполняют и методы `.ljust(n)` и `.rjust(n)` -
> обычно используются для формирования таблиц.

# 5. Индекс и срез: одно правило на всё

К символу строки обращаются по номеру в квадратных скобках.
**Нумерация начинается с нуля.** Отрицательные номера считаются с конца:
`[-1]` - последний.

Срез `[от:до]` принимает значения промежутка. **Правая граница не входит** - это правило
действует во всём Python, и привыкнуть к нему необходимо сразу.

```python
text = TEXTS[0]

print(text[0])
print(text[-1])

print(text[0:6])
print(text[:6])
print(text[-6:])
print(text[::2])
print(text[::-1])

print(text[:7] + text[7:] == text)
```

> **Срез никогда не выходит за границы.** `text[100:200]` вернёт пустую строку,
> а не ошибку. Обращение по индексу `text[100]` - это ошибка `IndexError`.
> То есть опечатка в срезе не падает, а молча возвращает пустоту, и найти
> её труднее.

```python
text = TEXTS[1]

print(repr(text[100:200]))
print(repr(text[10:5]))

# а это уже ошибка — раскомментируйте:
# print(text[100])
```

# 6. Из строки - в слова

Вопрос, с которого начинается любая обработка текста: **сколько здесь слов?**

**`.split()`** разделяет строку по пробелам и возвращает список.
**`.replace(что, на_что)`** заменяет; заменой на `""` удаляют.
**`.count(что)`** считает вхождения.

```python
text = TEXTS[0]

words = text.split()
print(words)
print(len(words))

print(len(text))
print(len(text.replace(" ", "")))

print(text.count("о"))
print(text.count("."))
```

```python
text = TEXTS[1]
print(text)
print(text.split())
print(TEXTS[0])
print(TEXTS[0].split()[-1])
```

`3:1` - это счёт матча, и `.split()` считает его словом.
А если бы запись кончалась точкой, последним «словом» было бы `океана.` -
со знаком.

Для Python слово - это просто то, что стоит между пробелами. **Что считать
словом на самом деле, решаете Вы**, и от этого решения меняется число
уникальных слов, частоты и всё, что на них построено. Никакая функция
не решит это за вас: это решение, а не вычисление.

# 7. Числа

`+`, `-`, `*`, `/`.

| Оператор | Что делает | Пример |
| --- | --- | --- |
| `/` | деление, **всегда** даёт `float` | `7 / 2` -> `3.5`, `4 / 2` -> `2.0` |
| `//` | целая часть от деления | `7 // 2` -> `3` |
| `%` | остаток | `7 % 2` -> `1` |
| `**` | степень | `2 ** 10` -> `1024` |

`//` и `%` нужны, чтобы найти середину списка и проверить чётность.
`round(x, n)` округляет до `n` знаков.

```python
print(7 / 2, 4 / 2, type(4 / 2))
print(7 // 2, 7 % 2, 2 ** 10)

text = TEXTS[0]
average = len(text.replace(" ", "")) / len(text.split())
print(average)
print(round(average, 2))
print(f"{average:.2f}")

# раскомментируйте:
# print(10 / 0)
```

> `round` возвращает **число**, а `f"{x:.2f}"` - **строку для печати**.
> Пока вы только печатаете, разницы не видно; она заметна, когда с результатом
> необходимо выполнить какую-либо операцию.
>
> И помните про деление на ноль: считая среднее, всегда проверяйте, что
> делить есть на что. Пустой список - самая частая причина `ZeroDivisionError`.

# 8. Список

Список - это несколько значений в квадратных скобках. Нумерация и срезы -
те же, что у строк, с тем же правилом про правую границу.

Списки в этом курсе идут **параллельно**: `TEXTS[1]`, `LABELS[1]` и `SCORES[1]`,
т.е. это одна и та же запись. Номер индекса связывает их между собой. Это самый простой
способ хранить таблицу, и Вы будете им пользоваться на первых этапах обучения.

```python
print(TEXTS)
print(LABELS)
print(SCORES)

print(len(TEXTS))
print(LABELS[0], LABELS[-1])
print(SCORES[:2])

i = 1
print(f"[{i}] {LABELS[i]}: {TEXTS[i]} (оценка {SCORES[i]})")
```

### Методы списка

`sum`, `min`, `max` работают со всем списком сразу. `in` проверяет наличие,
`.index()` находит номер первого вхождения, `.count()` считает повторы.

```python
print(sum(SCORES), min(SCORES), max(SCORES))
print(sum(SCORES) / len(SCORES))

print("наука" in LABELS)
print("погода" in LABELS)
print(LABELS.index("спорт"))
print(LABELS.count("наука"))
```

# 9. `sorted()` VS `.sort()`

Две вещи, которые легко перепутать:

- **`sorted(x)`** возвращает **новый** отсортированный список, исходный не трогает;
- **`x.sort()`** сортирует **на месте** и возвращает `None` - то есть ничего.

`x = my_list.sort()` кладёт в `x` пустоту. Это классическая ошибка,
и ошибки при этом не будет - будет `None` там, где вы ждали список.

```python
values = [3, 1, 2]

print(sorted(values))
print(values)

result = values.sort()
print(result)
print(values)
```

### И отдельно - про букву «Ё»

```python
print(sorted(["Ёмкость", "Арбуз", "Вышел"]))
print(ord("Ё"), ord("А"), ord("В"))
```

«Ёмкость» при сортировке переместилась на первое место.
Это не сбой: Python сравнивает строки по номерам символов в таблице **Unicode**,
а «Ё» стоит там под номером 1025 — **до** «А» (1040), отдельно от остального
алфавита.

Запомните этот случай. «Отсортировано» и «отсортировано так, как ожидает
человек» - немного разные вещи, и расходятся они почти всегда на букве «ё».

# 10. Список можно менять - и это опасно

В отличие от строки, список **изменяемый**: `.append(x)` добавляет элемент в конец списка,
`.insert(номер, x)` вставляет элемент в определенное место, `.remove(x)` удаляет первое вхождение элемента.

```python
values = [1, 2]
values.append(3)
values.insert(0, 0)
print(values)
values.remove(0)
print(values)
```

### Ловушка, которая стоит дороже всех остальных

Присваивание списка **не создаёт копию**. Два имени начинают указывать
на один и тот же список, и изменение через одно имя видно через другое.

```python
original = [1, 2, 3]
same = original
copy = original.copy()

same.append(999)

print("original:", original)
print("same:    ", same)
print("copy:    ", copy)
```

`original` изменился, хотя вы его не трогали. Ошибки здесь нет - есть неверные
данные, и обнаруживается это обычно далеко от места, где что-то пошло не по плану.

Правило простое: собираетесь изменять список, сначала нуобходимо создать копию списка при помощи метода `.copy()`.

# 11. Кортеж

Кортеж - тот же, что список, но находится он в круглых скобках, а главное его отличие - он **неизменяемый**.

Зачем он нужен, понятно не сразу, поэтому вот прямой ответ: список хранит
однотипное («десять оценок»), кортеж - **одну запись целиком**
(«метка, оценка, длина»). И кортеж умеет то, чего не умеет список:
распаковываться по переменным одной строкой.

```python
record = ("наука", 5, 58)

print(record)
print(record[0])

label, score, length = record
print(label, score, length)

a, b = 1, 2
a, b = b, a
print(a, b)

# раскомментируйте - кортеж изменить нельзя:
# record[1] = 3
```

### И ещё кортежи сортируются сами

Список кортежей сортируется **без всяких дополнительных указаний**:
кортежи сравниваются по первому элементу, при равенстве - по второму.

Отсюда приём, которым вы будете пользоваться постоянно: чтобы отсортировать
что-нибудь по числу, храните число в кортеже первым.

```python
pairs = [(5, "наука"), (3, "спорт"), (4, "наука")]

print(sorted(pairs))
print(sorted(pairs, reverse=True))
print(sorted(pairs)[-1])
print(pairs)
```

# 12. Множество

Множество хранит только **уникальные** значения - повторы исчезают самостоятельно.
Порядок в нём не сохраняется, поэтому его почти всегда оборачивают
в `sorted()`.

Два множества можно сравнивать между собой при помощи операторов: `&` - что есть в обоих,
`|` - всё вместе, `-` - что есть в первом и нет во втором.

```python
print(set(["кот", "пёс", "кот", "ёж", "кот"]))
print(len(set(LABELS)))

first = set(TEXTS[0].lower().split())
second = set(TEXTS[2].lower().split())

print("слов в первой:", len(first))
print("слов в третьей:", len(second))
print("общие:", sorted(first & second))
print("всего разных:", len(first | second))
print("только в первой:", len(first - second))
```

Доля общих слов - это простейшая мера похожести двух текстов.
Посчитайте её и посмотрите, из чего она складывается.

```python
first = set(TEXTS[0].lower().split())
second = set(TEXTS[2].lower().split())

similarity = len(first & second) / len(first | second)
print(f"похожесть: {similarity:.2f}")
print("а общие слова — вот эти:", sorted(first & second))
```

Оба текста про учёных и про открытия, а общих слов всего два, и одно
из них - это предлог «в».

Это и есть главная трудность обработки текста, ради которой существует весь
остальной курс: **буквальное совпадение слов почти ничего не говорит
о смысле**.

# Шпаргалка занятия

| Что нужно | Как пишется | Замечание |
| --- | --- | --- |
| напечатать | `print(a, b)` | между значениями встанет пробел |
| комментарий | `# текст` | |
| положить в переменную | `x = значение` | «положи в», не «равно» |
| тип | `type(x)` | `str`, `int`, `float`, `bool` |
| длина | `len(x)` | строка, список, множество |
| регистр | `s.upper()`, `s.lower()` | возвращают новую строку |
| склеить | `a + b` | только строку со строкой |
| собрать текст | `f"{a} и {b}"` | типы не важны |
| формат числа | `f"{x:.2f}"`, `f"{x:.0%}"` | результат — строка |
| столбик | `.ljust(n)`, `.rjust(n)` | |
| символ / элемент | `x[0]`, `x[-1]` | нумерация с нуля |
| срез | `x[2:7]`, `x[:3]`, `x[-2:]` | правая граница не входит |
| в слова | `s.split()` | по пробелам |
| заменить / удалить | `s.replace(a, b)` | |
| посчитать вхождения | `s.count(a)` | |
| деление / целое / остаток | `/`, `//`, `%` | `/` всегда даёт `float` |
| округлить | `round(x, 2)` | результат — число |
| сумма, минимум, максимум | `sum`, `min`, `max` | для списка чисел |
| есть ли | `x in список` | |
| номер первого | `список.index(x)` | |
| сортировка | `sorted(x)` — новый список | `x.sort()` — на месте, вернёт `None` |
| изменить список | `.append()`, `.insert()`, `.remove()` | |

| распаковка | `a, b, c = кортеж` | |

# Частые ошибки первого занятия

Здесь собрано то, на чём спотыкаются почти все. Если что-то в практике
не работает, скорее всего это одно из перечисленного.

**1. Забыты скобки у метода.** `text.upper` выведет описание метода,
а не результат. Ошибки нет.

**2. Перепутаны `=` и `==`.** Одинарное присваивает значение, двойное сравнивает.

**3. Склеили строку с числом.** `"оценка: " + 5` — `TypeError`.
Нужна f-строка.

**4. Ожидали `2`, получили `2.0`.** Деление всегда даёт `float`.

**5. Обратились к `x[len(x)]`.** Нумерация с нуля, последний номер -
`len(x) - 1`. Или просто `x[-1]`.

**6. `x = my_list.sort()`** — в `x` оказался `None`. Нужен `sorted()`.

**7. Присвоили список другому имени и изменили** - изменились оба.
Нужен `.copy()`.

**8. Строка не находится, хотя она есть** - не совпал регистр.
Перед поиском приводите к `.lower()`.

**9. Срез вернул пустоту вместо ошибки** - границы перепутаны местами
или вышли за длину. Проверьте `len()`.

**10. Блокнот ведёт себя странно после правок** - ячейки выполнялись
не по порядку. *Kernel → Restart & Run All*.
