`round(число, знаков)` округляет. В f-строке того же можно добиться записью

## Задача 27. Общие слова двух новостей

*Новое: `&`, `|`, `-` у множеств*

Два множества можно сравнить:

- `a & b` - что есть **в обоих** (пересечение);
- `a | b` - всё вместе, без повторов (объединение);
- `a - b` - что есть в `a`, но нет в `b` (разность).

Это первый инструмент, которым сравнивают тексты.

````python
left = set("кот спит на окне".split())
right = set("собака спит на коврике".split())

print(sorted(left & right))
print(len(left | right))
print(sorted(left - right))
````
